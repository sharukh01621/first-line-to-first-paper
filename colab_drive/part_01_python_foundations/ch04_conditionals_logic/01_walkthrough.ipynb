## Section 4.3.1 — Comparison Operators & Chained Clinical Ranges

```python
sbp_mmhg = 114
hb_g_dl = 12.2
temp_c = 38.4

# Chained comparisons mirror clinical reference intervals directly
is_normal_bp = 90 <= sbp_mmhg < 120
is_normal_hb = 11.0 <= hb_g_dl <= 16.0
has_fever = temp_c >= 38.0

print("Normal BP:", is_normal_bp)
print("Normal Hb:", is_normal_hb)
print("Has Fever:", has_fever)
```

## Section 4.3.2 — Boolean Algebra (`and`, `or`, `not`) & Operator Precedence

```python
has_fever = True
has_rash = True
is_vaccinated = False

# Precedence: not > and > or (parentheses make clinical intent explicit!)
suspect_measles = has_fever and has_rash and (not is_vaccinated)
needs_isolation = suspect_measles or (has_fever and not is_vaccinated)

print("Suspect measles:", suspect_measles)
print("Needs isolation:", needs_isolation)
```

## Section 4.3.3 — `if` / `elif` / `else` Clinical Decision Tree (and the `if` vs. `elif` Bug!)

```python
# Correct WHO maternal anemia severity ladder (mutually exclusive elif)
hb_g_dl = 6.4

if hb_g_dl < 7.0:
    severity = "Severe"
elif hb_g_dl < 10.0:
    severity = "Moderate"
elif hb_g_dl < 11.0:
    severity = "Mild"
else:
    severity = "Normal"

print(f"Hb {hb_g_dl} g/dL -> Correct severity: {severity}")
```

```python
# Silent Logic Bug: independent 'if' statements overwrite earlier matches!
hb_g_dl = 6.4
buggy_severity = "Normal"

if hb_g_dl < 7.0:
    buggy_severity = "Severe"
if hb_g_dl < 10.0:
    buggy_severity = "Moderate"
if hb_g_dl < 11.0:
    buggy_severity = "Mild"   # DANGEROUS: overwrites 'Severe' with 'Mild'!

print(f"Hb {hb_g_dl} g/dL -> Buggy severity:   {buggy_severity}")
```

## Section 4.3.4 — Short-Circuit Evaluation & Guard Clauses (`population > 0`, `None`, `999`)

```python
# Guard 1: Short-circuit prevents ZeroDivisionError when population == 0
cases = 18
population = 0

high_outbreak = population > 0 and (cases / population) * 100000 >= 50.0
print("High outbreak (safely guarded):", high_outbreak)

# Guard 2: Missing-value guard intercepts BOTH None and DHS sentinel 999
raw_hb = 999
if raw_hb is None or raw_hb == 999:
    hb_label = "Missing"
elif raw_hb < 11.0:
    hb_label = "Anemic"
else:
    hb_label = "Normal"

print("Guarded Hb status:", hb_label)
```

## Section 4.4 — Worked Problem-Structuring Blueprint: Upazila Maternal Triage Router

```python
# Step 1: Patient intake readings from Upazila Health Complex
hb_g_dl = 6.8
sbp_mmhg = 152
has_headache = True

# Step 2: Missing-value guard + WHO maternal anemia classification
if hb_g_dl is None or hb_g_dl == 999:
    anemia_grade = "Missing"
elif hb_g_dl < 7.0:
    anemia_grade = "Severe"
elif hb_g_dl < 10.0:
    anemia_grade = "Moderate"
elif hb_g_dl < 11.0:
    anemia_grade = "Mild"
else:
    anemia_grade = "Normal"

# Step 3: Preeclampsia warning & clinic triage routing
preeclampsia_flag = (sbp_mmhg >= 140) and has_headache

if anemia_grade == "Severe" or sbp_mmhg >= 160 or preeclampsia_flag:
    triage_priority = "Urgent"
elif anemia_grade == "Moderate" or sbp_mmhg >= 140:
    triage_priority = "Moderate"
else:
    triage_priority = "Routine"

# Step 4: Structured clinical triage readout
print(f"Anemia: {anemia_grade} | Preeclampsia Flag: {preeclampsia_flag}")
print(f"Upazila Triage Priority: {triage_priority}")
```
