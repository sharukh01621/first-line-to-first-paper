## Exam Task 20.1 (30 Points): `epidemiological_2x2_report(a, b, c, d)`

Given 2x2 contingency counts (`a`: exposed cases, `b`: exposed non-cases, `c`: unexposed cases, `d`: unexposed non-cases), compute and return a single dictionary with exact keys:
1. `"risk_exposed"`: $a / (a + b)$ rounded to **4 decimals**.
2. `"risk_unexposed"`: $c / (c + d)$ rounded to **4 decimals**.
3. `"risk_ratio"`: $\text{risk\_exposed} / \text{risk\_unexposed}$ rounded to **3 decimals**.
4. `"odds_ratio"`: $(a \cdot d) / (b \cdot c)$ rounded to **3 decimals**.
5. `"or_ci_low"` and `"or_ci_high"`: Woolf 95% CI ($z = 1.96$) rounded to **3 decimals**.
6. `"chi2_p_value"`: Pearson chi-square $p$-value from `stats.chi2_contingency([[a, b], [c, d]], correction=False)` rounded to **4 decimals**.

```python
def epidemiological_2x2_report(a, b, c, d):
    """Return full 2x2 risk, RR, OR, Woolf 95% CI, and chi2 p-value."""
    risk_exp = a / (a + b)
    risk_unexp = c / (c + d)
    rr = risk_exp / risk_unexp
    or_val = (a * d) / (b * c)
    se = math.sqrt(1 / a + 1 / b + 1 / c + 1 / d)
    log_or = math.log(or_val)
    _, p_val, _, _ = stats.chi2_contingency(
        [[a, b], [c, d]], correction=False
    )
    return {
        "risk_exposed": round(risk_exp, 4),
        "risk_unexposed": round(risk_unexp, 4),
        "risk_ratio": round(rr, 3),
        "odds_ratio": round(or_val, 3),
        "or_ci_low": round(math.exp(log_or - 1.96 * se), 3),
        "or_ci_high": round(math.exp(log_or + 1.96 * se), 3),
        "chi2_p_value": round(float(p_val), 4),
    }
```

## Exam Task 20.2 (35 Points): `anova_and_kruskal_by_group(df, value_col, group_col)`

Compare a continuous outcome across $k \ge 2$ groups both parametrically and nonparametrically without mutating `df`:
1. Group `df` by `group_col` and extract non-null values in `value_col` per group (dropping `NaN`s and skipping empty groups).
2. Run `stats.f_oneway(*groups)` (One-Way ANOVA) and `stats.kruskal(*groups)` (Kruskal-Wallis $H$ test).
3. Return `{"n_groups": int, "f_stat": round4, "anova_p": round4, "h_stat": round4, "kruskal_p": round4}`.

```python
def anova_and_kruskal_by_group(df, value_col, group_col):
    """Run NaN-safe One-Way ANOVA and Kruskal-Wallis across groups."""
    groups = [
        grp[value_col].dropna().to_numpy()
        for _, grp in df.groupby(group_col)
        if grp[value_col].notna().any()
    ]
    f_stat, anova_p = stats.f_oneway(*groups)
    h_stat, kruskal_p = stats.kruskal(*groups)
    return {
        "n_groups": int(len(groups)),
        "f_stat": round(float(f_stat), 4),
        "anova_p": round(float(anova_p), 4),
        "h_stat": round(float(h_stat), 4),
        "kruskal_p": round(float(kruskal_p), 4),
    }
```

## Exam Task 20.3 (35 Points): `fit_adjusted_anemia_logit(mch_df)`

Build a publication-ready multivariable logistic regression pipeline (without mutating `mch_df`):
1. Copy `mch_df`, standardize `residence` casing (`str.strip().str.title()`), and replace sentinel `anc_visits == 999` with `np.nan`.
2. Create binary `is_urban = (work["residence"] == "Urban").astype(int)`.
3. Drop rows with missing values in `["maternal_anemia", "anc_visits", "is_urban"]`.
4. Fit `sm.Logit(y, sm.add_constant(X)).fit(disp=False)` predicting `maternal_anemia` from `anc_visits` and `is_urban`.
5. Return a `pd.DataFrame` indexed by `["const", "anc_visits", "is_urban"]` with columns `["aor", "ci_low", "ci_high", "p_value"]` (exponentiated parameters & 95% CIs, plus p-values) rounded to **4 decimal places**.

```python
def fit_adjusted_anemia_logit(mch_df):
    """Clean MCH survey covariates and fit multivariable Logit aORs."""
    work = mch_df.copy()
    work["residence"] = (
        work["residence"].astype(str).str.strip().str.title()
    )
    work["anc_visits"] = work["anc_visits"].replace(999, np.nan)
    work["is_urban"] = (work["residence"] == "Urban").astype(int)
    model_df = work[
        ["maternal_anemia", "anc_visits", "is_urban"]
    ].dropna()
    X = sm.add_constant(model_df[["anc_visits", "is_urban"]])
    y = model_df["maternal_anemia"]
    fit = sm.Logit(y, X).fit(disp=False)
    conf = fit.conf_int()
    return pd.DataFrame(
        {
            "aor": np.exp(fit.params),
            "ci_low": np.exp(conf[0]),
            "ci_high": np.exp(conf[1]),
            "p_value": fit.pvalues,
        },
        index=["const", "anc_visits", "is_urban"],
    ).round(4)
```

## Part IV Capstone Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 20.1 (RR vs. OR Divergence & Yates' Correction):** When baseline risk is non-trivial (e.g., 25%–70%), the Odds Ratio (`7.000`) dramatically exaggerates the Risk Ratio (`2.800`). Always compute both from exact raw probabilities before rounding, and pass `correction=False` to `stats.chi2_contingency` for standard Pearson $\chi^2$.
- **Task 20.2 (NaN Poisoning in `f_oneway` and `kruskal`):** Passing even one `NaN` inside any group array causes both `stats.f_oneway` and `stats.kruskal` to propagate `nan` across the entire statistic and p-value. Calling `.dropna()` per group prevents silent statistical corruption.
- **Task 20.3 (Sentinel `999`, Intercept & `np.exp`):** Leaving `anc_visits == 999` uncleaned crushes the logistic slope toward zero and wrecks the aOR! Remember three essential Statsmodels steps: (1) clean sentinels and drop NaNs on a `.copy()`, (2) include `sm.add_constant(X)` so the intercept `const` is estimated, and (3) exponentiate log-odds `fit.params` and `fit.conf_int()` with `np.exp()` to obtain interpretable Adjusted Odds Ratios.
