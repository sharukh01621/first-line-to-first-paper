## To-Do 17.1 (Vectorized Column & Copy-on-Write): `add_bmi_column(df)`

Implement `add_bmi_column(df)`:
- Return a **fresh copy** of `df` with a new column `'bmi'` computed as `(weight_kg / (height_cm / 100) ** 2).round(2)`.
- **Never mutate** the caller's input DataFrame `df` in place.

```python
def add_bmi_column(df):
    """Return a fresh copy of df with 2-decimal BMI column added."""
    out = df.copy()
    height_m = out["height_cm"] / 100
    out["bmi"] = (out["weight_kg"] / (height_m ** 2)).round(2)
    return out
```

## To-Do 17.2 (Compound Boolean Filtering): `filter_urgent_adults(df)`

Implement `filter_urgent_adults(df)`:
- Keep rows where `age_years >= 18` **and** `triage_priority == "Urgent"`.
- Reset the index with `drop=True` and return a fresh copy without mutating `df`.

```python
def filter_urgent_adults(df):
    """Filter adult Urgent rows, reset index, and return a copy."""
    mask = (df["age_years"] >= 18) & (df["triage_priority"] == "Urgent")
    return df.loc[mask].reset_index(drop=True).copy()
```

## To-Do 17.3 (Multi-Key Sorting & Projection): `select_top_fever_patients(df, n=5)`

Implement `select_top_fever_patients(df, n=5)`:
- Sort by `temp_c` descending, breaking ties by `patient_id` ascending.
- Return the top `n` rows with exact columns `["patient_id", "district", "temp_c"]` and a reset index (`drop=True`), without mutating `df`.

```python
def select_top_fever_patients(df, n=5):
    """Return top-n fever patients sorted by temp_c desc, id asc."""
    cols = ["patient_id", "district", "temp_c"]
    ranked = df.sort_values(
        by=["temp_c", "patient_id"],
        ascending=[False, True],
    )
    return ranked[cols].head(n).reset_index(drop=True)
```

```python
tutor.check("ch17_todo_03", select_top_fever_patients)
```

## To-Do 17.4 (Numeric Series Summary Dict): `summarize_numeric_vitals(df, cols)`

Implement `summarize_numeric_vitals(df, cols)`:
- Return a dictionary mapping each column name `col` in `cols` to `{"mean": float, "median": float, "std": float}`.
- Round every statistic to **2 decimal places** using Pandas' default sample standard deviation (`ddof=1`).

```python
def summarize_numeric_vitals(df, cols):
    """Return {col: {'mean', 'median', 'std'}} rounded to 2 decimals."""
    summary = {}
    for col in cols:
        s = df[col]
        summary[col] = {
            "mean": float(round(s.mean(), 2)),
            "median": float(round(s.median(), 2)),
            "std": float(round(s.std(), 2)),
        }
    return summary
```

```python
tutor.check("ch17_todo_04", summarize_numeric_vitals)
```

## To-Do 17.5 (Wide-to-Long Reshaping): `melt_vital_measurements(wide_df)`

Implement `melt_vital_measurements(wide_df)`:
- Given a wide DataFrame with columns `["patient_id", "sbp_day1", "sbp_day2"]`, use `pd.melt` to reshape it into tidy long format with exact columns `["patient_id", "visit", "sbp"]`.

```python
def melt_vital_measurements(wide_df):
    """Unpivot sbp_day1/sbp_day2 to long [patient_id, visit, sbp]."""
    return pd.melt(
        wide_df,
        id_vars=["patient_id"],
        value_vars=["sbp_day1", "sbp_day2"],
        var_name="visit",
        value_name="sbp",
    )
```

## To-Do 17.6 (Contingency Table): `pivot_district_priority_counts(df)`

Implement `pivot_district_priority_counts(df)`:
- Return a contingency DataFrame with `district` as the index, `triage_priority` as columns, and **integer counts** with missing combinations filled with `0`.

```python
def pivot_district_priority_counts(df):
    """Return integer district x triage_priority counts filled with 0."""
    return pd.crosstab(df["district"], df["triage_priority"])
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 17.1 (Non-Mutation & Copy-on-Write):** Never write `df["bmi"] = ...` directly on the caller's DataFrame inside a helper function—always start with `out = df.copy()` or `df.assign(...)` so upstream tables remain reproducible.
- **To-Do 17.2 (Bitwise `&` & `drop=True`):** Python's `and` fails on boolean Series (`ValueError: The truth value of a Series is ambiguous`), and calling `.reset_index()` without `drop=True` injects an unwanted `'index'` column.
- **To-Do 17.4 (`ddof=1` vs. `ddof=0`):** Pandas `Series.std()` computes **sample** standard deviation (`ddof=1`), whereas NumPy `np.std()` defaults to **population** (`ddof=0`). Wrapping with `float(round(..., 2))` also ensures clean native Python floats for JSON serialization.
