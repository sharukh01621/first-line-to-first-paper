## Exam Task 1.1 (30 Points): Pediatric ORS Supply & Carton Logistics

A community clinic projects `140` pediatric diarrhea patients this week. Each patient requires `3` Oral Rehydration Solution (ORS) packets, and district guidelines add a **15% emergency safety buffer** (`* 1.15`). Each shipping carton holds `50` packets.

**Problem Blueprint:**
- **Inputs:** `patients = 140`, `packets_per_patient = 3`, `buffer_multiplier = 1.15`, `carton_size = 50`
- **Outputs to compute:**
  1. `base_packets` (int): total packets before safety buffer
  2. `total_packets` (int): total packets including the 15% buffer (use `round(base_packets * buffer_multiplier)` so binary floating-point rounding yields the exact integer `483`)
  3. `full_cartons` (int): number of completely full 50-packet cartons
  4. `extra_packets` (int): remaining loose packets

```python
patients = 140
packets_per_patient = 3
buffer_multiplier = 1.15
carton_size = 50

base_packets = patients * packets_per_patient
total_packets = round(base_packets * buffer_multiplier)
full_cartons = total_packets // carton_size
extra_packets = total_packets % carton_size

print(f"Base: {base_packets}, With 15% buffer: {total_packets} -> {full_cartons} cartons + {extra_packets} extra")
```

## Exam Task 1.2 (35 Points): Maternal Health Clinic Summary Card

During Q1, `320` pregnant mothers registered at a rural Maternal & Child Health center:
- `218` completed 4 or more antenatal care (ANC 4+) visits.
- `96` tested positive for maternal anemia ($\text{Hb} < 11.0\text{ g/dL}$).

**Outputs to compute:**
1. `anc_coverage_pct`: percentage completing 4+ ANC visits, rounded to **1 decimal place**
2. `anemia_prevalence_pct`: percentage with anemia, rounded to **1 decimal place**
3. `summary_line`: formatted string matching `"ANC 4+ Coverage: 68.1% | Anemia Prevalence: 30.0%"` using your computed variables

```python
total_mothers = 320
anc_4plus = 218
anemic_count = 96

anc_coverage_pct = round((anc_4plus / total_mothers) * 100, 1)
anemia_prevalence_pct = round((anemic_count / total_mothers) * 100, 1)
summary_line = f"ANC 4+ Coverage: {anc_coverage_pct}% | Anemia Prevalence: {anemia_prevalence_pct}%"

print(summary_line)
```

## Exam Task 1.3 (35 Points): Rural vs. Urban Case Fatality Disparity

Surveillance reports for a seasonal outbreak show:
- **Rural District:** `450` confirmed cases, `9` deaths
- **Urban District:** `600` confirmed cases, `6` deaths

Recall: $\text{Case Fatality Rate (\%)} = \left(\frac{\text{deaths}}{\text{confirmed cases}}\right) \times 100$.

**Outputs to compute:**
1. `cfr_rural_pct`: Case Fatality Rate (%) in the rural district
2. `cfr_urban_pct`: Case Fatality Rate (%) in the urban district
3. `cfr_ratio`: Relative ratio `cfr_rural_pct / cfr_urban_pct`

```python
rural_cases = 450
rural_deaths = 9
urban_cases = 600
urban_deaths = 6

cfr_rural_pct = (rural_deaths / rural_cases) * 100
cfr_urban_pct = (urban_deaths / urban_cases) * 100
cfr_ratio = cfr_rural_pct / cfr_urban_pct

print(f"Rural CFR: {cfr_rural_pct}% | Urban CFR: {cfr_urban_pct}% | Disparity Ratio: {cfr_ratio}x")
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 1.1 (Famous Binary Floating-Point Trap!):** In IEEE-754 binary math, `420 * 1.15` evaluates to `482.99999999999994`, NOT `483.0`! Using `int(420 * 1.15)` truncates downward to `482` and loses a packet! Using `round(base_packets * buffer_multiplier)` gives the exact integer `483`, yielding `9` full cartons and `33` extra packets. We unpack *why* binary floats behave this way in Chapter 2.
- **Task 1.2:** Hardcoding the text `"68.1%"` instead of using an f-string breaks reproducibility whenever next quarter's clinic numbers arrive. Always inject computed variables into report strings.
- **Task 1.3:** Inverting the ratio (`urban / rural = 0.5`) reverses the epidemiological narrative! Always state the reference group explicitly (here, Urban is the denominator/reference).
