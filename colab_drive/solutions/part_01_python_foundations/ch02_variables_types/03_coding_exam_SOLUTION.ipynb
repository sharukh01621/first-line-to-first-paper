## Exam Task 2.1 (30 Points): DHS Survey Record Normalization

A raw Maternal & Child Health survey record contains:
`raw_weight_kg = "54.6"`, `raw_height_cm = "156.0"`, `raw_anc_visits = "4.0"`.

Compute with exact required types:
1. `weight_kg` (`float`)
2. `height_m` (`float`, converted from centimeters to meters)
3. `anc_visits` (`int`, parsed safely from `"4.0"`)
4. `bmi` (`float`, rounded to 2 decimal places)

```python
raw_weight_kg = "54.6"
raw_height_cm = "156.0"
raw_anc_visits = "4.0"

weight_kg = float(raw_weight_kg)
height_m = float(raw_height_cm) / 100
anc_visits = int(float(raw_anc_visits))
bmi = round(weight_kg / (height_m ** 2), 2)

print(f"Weight: {weight_kg} kg | Height: {height_m} m | ANC: {anc_visits} | BMI: {bmi}")
```

## Exam Task 2.2 (35 Points): 4-Day Outbreak Accumulator & Threshold Alert

A district of `population = 250000` reports daily dengue counts over 4 consecutive days: `18`, `27`, `34`, `21`.

1. Start `total_cases = 0` and accumulate all 4 days using `+=` (`int`).
2. Compute `mean_daily_cases` (`total_cases / 4`).
3. Compute `incidence_per_100k = (total_cases / population) * 100000`.
4. Compute boolean `outbreak_alert`: `True` if `incidence_per_100k >= 35.0`.

```python
population = 250000
day_1, day_2, day_3, day_4 = 18, 27, 34, 21

total_cases = 0
total_cases += day_1
total_cases += day_2
total_cases += day_3
total_cases += day_4
mean_daily_cases = total_cases / 4
incidence_per_100k = (total_cases / population) * 100000
outbreak_alert = incidence_per_100k >= 35.0

print(f"Total: {total_cases} | Daily mean: {mean_daily_cases} | Rate/100k: {incidence_per_100k} | Alert: {outbreak_alert}")
```

## Exam Task 2.3 (35 Points): Biostatistics 2x2 Table Odds & Odds Ratio

In a maternal anemia study comparing mothers who did **not** receive iron supplementation (Exposed) vs. supplemented mothers (Unexposed):
- Exposed (no iron): `a = 45` anemic, `b = 105` non-anemic
- Unexposed (iron): `c = 20` anemic, `d = 130` non-anemic

Compute:
1. `odds_exposed = round(a / b, 4)`
2. `odds_unexposed = round(c / d, 4)`
3. `odds_ratio = round((a * d) / (b * c), 2)` *(exact cross-product formula before rounding!)*
4. `elevated_risk = odds_ratio > 1.0` (`bool`)

```python
a, b = 45, 105
c, d = 20, 130

odds_exposed = round(a / b, 4)
odds_unexposed = round(c / d, 4)
odds_ratio = round((a * d) / (b * c), 2)
elevated_risk = odds_ratio > 1.0

print(f"Odds(Exposed): {odds_exposed} | Odds(Unexposed): {odds_unexposed} | OR: {odds_ratio} | Elevated: {elevated_risk}")
```
