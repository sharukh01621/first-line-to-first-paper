## To-Do 6.1 (Warm-Up): Celsius to Fahrenheit Converter

Write a pure function `celsius_to_fahrenheit(temp_c)` that converts a temperature from Celsius to Fahrenheit using `(temp_c * 9 / 5) + 32` and **returns** the result rounded to **2 decimal places**.

*Remember:* Always `return` the value — do NOT just `print()` it!

```python
def celsius_to_fahrenheit(temp_c):
    """Convert Celsius temperature to Fahrenheit, rounded to 2 decimals."""
    return round((temp_c * 9 / 5) + 32, 2)
```

## To-Do 6.2 (Unit Conversion Inside a Function): Reusable BMI Calculator

Write `compute_bmi(weight_kg, height_cm)` that converts `height_cm` to meters (`height_cm / 100`), computes `weight_kg / (height_m ** 2)`, and **returns** the BMI rounded to **2 decimal places**.

```python
def compute_bmi(weight_kg, height_cm):
    """Return Body Mass Index (kg/m^2) rounded to 2 decimal places."""
    height_m = height_cm / 100
    return round(weight_kg / (height_m ** 2), 2)
```

## To-Do 6.3 (Default Parameters): WHO Anemia Classifier

Write `classify_anemia(hb_g_dl, pregnant=True)` using WHO hemoglobin thresholds:
- When `pregnant=True` (default): return `"Anemic"` if `hb_g_dl < 11.0`, else `"Normal"`.
- When `pregnant=False`: return `"Anemic"` if `hb_g_dl < 12.0`, else `"Normal"`.

```python
def classify_anemia(hb_g_dl, pregnant=True):
    """Classify anemia using WHO cutoffs (11.0 pregnant, 12.0 other)."""
    cutoff = 11.0 if pregnant else 12.0
    if hb_g_dl < cutoff:
        return "Anemic"
    return "Normal"
```

## To-Do 6.4 (Guard Clauses + Defaults): Safe Incidence Rate

Write `safe_incidence_rate(cases, population, per=100000)`:
- If `population <= 0`, return `None` immediately (guard clause preventing `ZeroDivisionError`).
- Otherwise return `round((cases / population) * per, 2)`.

```python
def safe_incidence_rate(cases, population, per=100000):
    """Return incidence rate per N people, or None if population <= 0."""
    if population <= 0:
        return None
    return round((cases / population) * per, 2)
```

```python
tutor.check("ch06_todo_04", safe_incidence_rate)
```

## To-Do 6.5 (Sentinel Filtering Helper): Clean Mean of Lab Readings

Write `clean_mean_readings(readings)` that accepts a list of lab readings:
- Skip any `None` entries and any sentinel `999` values.
- If no valid readings remain, return `None` (no `ZeroDivisionError`!).
- Otherwise return the arithmetic mean of valid readings rounded to **2 decimal places**.

```python
def clean_mean_readings(readings):
    """Return mean of valid readings (skipping None and 999), or None."""
    total = 0.0
    valid_count = 0
    for val in readings:
        if val is None or val == 999:
            continue
        total += val
        valid_count += 1
    if valid_count == 0:
        return None
    return round(total / valid_count, 2)
```

```python
tutor.check("ch06_todo_05", clean_mean_readings)
```

## To-Do 6.6 (Stretch Clinical Logic): Multi-Indicator Triage Priority

Write `triage_priority(temp_c, sbp_mmhg, hb_g_dl)` returning one of three priority labels:
- `"Urgent"` if `temp_c >= 39.5` **or** `sbp_mmhg >= 160` **or** `hb_g_dl < 8.0`
- `"Moderate"` if `temp_c >= 38.0` **or** `sbp_mmhg >= 140` **or** `hb_g_dl < 11.0`
- `"Routine"` otherwise.

```python
def triage_priority(temp_c, sbp_mmhg, hb_g_dl):
    """Return 'Urgent', 'Moderate', or 'Routine' from clinical vitals."""
    if temp_c >= 39.5 or sbp_mmhg >= 160 or hb_g_dl < 8.0:
        return "Urgent"
    if temp_c >= 38.0 or sbp_mmhg >= 140 or hb_g_dl < 11.0:
        return "Moderate"
    return "Routine"
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **The `print()` vs. `return` Trap (To-Dos 6.1–6.6):** Writing `print(result)` instead of `return result` makes a function silently return `None`, causing downstream arithmetic or `tutor.check(...)` assertions to fail immediately.
- **To-Do 6.4 & 6.5 (Guard Clauses):** Putting `if population <= 0: return None` and `if valid_count == 0: return None` before division eliminates `ZeroDivisionError` crashes when processing imperfect field registries.
- **To-Do 6.6 (Severity Ordering):** Checking `"Moderate"` (`temp_c >= 38.0`) before `"Urgent"` (`temp_c >= 39.5`) would mistakenly route a `39.8 C` critical patient into the `"Moderate"` queue—always test highest severity first!
