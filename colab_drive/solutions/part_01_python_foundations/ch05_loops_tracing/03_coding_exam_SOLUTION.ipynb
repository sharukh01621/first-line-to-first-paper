## Exam Task 5.1 (30 Points): Dirty Maternal Hemoglobin Cleaning & Summary

A maternal nutrition camp recorded 10 raw hemoglobin readings (in g/dL), where failed assays were coded as `999` or `None`:
`raw_hb_readings = [10.5, 999, 11.8, None, 9.2, 12.4, 999, 10.1, None, 11.0]`.

Use a `for` loop with `continue` to skip `None` and `999`, and compute:
1. `valid_count` (`int`): count of valid readings
2. `missing_count` (`int`): count of missing (`None` or `999`) readings
3. `mean_hb` (`float`): mean of valid readings rounded to **2 decimal places** (guarding `valid_count == 0`)
4. `anemic_count` (`int`): count of valid readings strictly `< 11.0` g/dL

```python
raw_hb_readings = [10.5, 999, 11.8, None, 9.2, 12.4, 999, 10.1, None, 11.0]

valid_count = 0
missing_count = 0
hb_sum = 0.0
anemic_count = 0

for hb in raw_hb_readings:
    if hb is None or hb == 999:
        missing_count += 1
        continue
    hb_sum += hb
    valid_count += 1
    if hb < 11.0:
        anemic_count += 1

if valid_count > 0:
    mean_hb = round(hb_sum / valid_count, 2)
else:
    mean_hb = None

print(f"Valid: {valid_count} | Missing: {missing_count} | Mean: {mean_hb} | Anemic: {anemic_count}")
```

## Exam Task 5.2 (35 Points): Monsoon Dengue Peak-Week & High-Alert Surveillance

A district reports 8 consecutive weeks of monsoon dengue cases:
`weekly_dengue = [42, 65, 98, 134, 178, 162, 115, 74]`  
with `high_alert_threshold = 100`.

Use `enumerate(weekly_dengue, start=1)` to compute:
1. `total_cases` (`int`): cumulative 8-week cases
2. `peak_week` (`int`): 1-based week number with the highest case count
3. `peak_cases` (`int`): highest weekly case count
4. `high_alert_weeks` (`int`): count of weeks with `cases >= 100`
5. `first_alert_week` (`int`): 1-based week number of the **first** week with `cases >= 100`

```python
weekly_dengue = [42, 65, 98, 134, 178, 162, 115, 74]
high_alert_threshold = 100

total_cases = 0
peak_week = 1
peak_cases = weekly_dengue[0]
high_alert_weeks = 0
first_alert_week = None

for week, cases in enumerate(weekly_dengue, start=1):
    total_cases += cases
    if cases > peak_cases:
        peak_cases = cases
        peak_week = week
    if cases >= high_alert_threshold:
        high_alert_weeks += 1
        if first_alert_week is None:
            first_alert_week = week

print(f"Total: {total_cases} | Peak: W{peak_week} ({peak_cases}) | Alerts: {high_alert_weeks} (First: W{first_alert_week})")
```

## Exam Task 5.3 (35 Points): Hospital Surge-Capacity Day Simulation (`while` Loop)

A district hospital begins an outbreak on Day `0` with:
- `occupied_beds = 40`
- `bed_capacity = 250`
- `daily_admissions = 20` (admissions on Day 1)
- `daily_discharges = 10` (constant daily recoveries)

Write a `while occupied_beds <= bed_capacity:` simulation where each day executes in exact order:
1. Increment `surge_day += 1`.
2. Update `occupied_beds += daily_admissions - daily_discharges`.
3. Accumulate `total_admitted += daily_admissions`.
4. Increase `daily_admissions += 5` for the following day.

After the loop exits, set `final_occupied_beds = occupied_beds` and `overflow_beds = occupied_beds - bed_capacity`.

```python
occupied_beds = 40
bed_capacity = 250
daily_admissions = 20
daily_discharges = 10

surge_day = 0
total_admitted = 0

while occupied_beds <= bed_capacity:
    surge_day += 1
    occupied_beds += daily_admissions - daily_discharges
    total_admitted += daily_admissions
    daily_admissions += 5

final_occupied_beds = occupied_beds
overflow_beds = occupied_beds - bed_capacity

print(f"Surge Day: {surge_day} | Beds: {final_occupied_beds} | Overflow: {overflow_beds} | Admitted: {total_admitted}")
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 5.1 (Boundary `< 11.0` & Missing Guard):** WHO defines maternal anemia as strictly `< 11.0` g/dL, so `11.0` is non-anemic (`anemic_count = 3`, not `4`). Guarding `if valid_count > 0:` before dividing prevents `ZeroDivisionError` on completely missing batches.
- **Task 5.2 (First-Occurrence Latch):** Setting `first_alert_week = week` unconditionally inside `if cases >= 100:` overwrites Week 4 with Week 7! Adding `if first_alert_week is None:` latches onto the very first qualifying week while still letting the loop finish summing all 8 weeks.
- **Task 5.3 (Statement Ordering in `while` Loops):** Moving `daily_admissions += 5` above `total_admitted += daily_admissions` silently inflates `total_admitted` from `300` to `340` because Day 1 counts `25` admissions instead of `20`. Always trace the first two iterations on paper!
