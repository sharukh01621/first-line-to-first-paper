## To-Do 5.1 (Warm-Up — Accumulator & Counter): 7-Day Vaccine Cold-Chain Tally

A rural EPI clinic recorded daily vaccine doses administered over one week:
`daily_doses = [45, 52, 48, 60, 55, 40, 50]`.

Use a `for` loop to compute:
- `total_doses` (`int`): sum of all 7 days
- `high_volume_days` (`int`): count of days with `>= 50` doses

```python
daily_doses = [45, 52, 48, 60, 55, 40, 50]

total_doses = 0
high_volume_days = 0
for doses in daily_doses:
    total_doses += doses
    if doses >= 50:
        high_volume_days += 1

print(f"Total doses: {total_doses} | High-volume days (>= 50): {high_volume_days}")
```

## To-Do 5.2 (Filter Pattern): Filtering Febrile Patients

Triage recorded body temperatures (in °C) for six incoming patients:
`temps_c = [36.8, 38.4, 37.5, 39.1, 38.0, 36.9]`.

Build a new list `febrile_temps` containing only temperatures `>= 38.0` (preserving original order).

```python
temps_c = [36.8, 38.4, 37.5, 39.1, 38.0, 36.9]

febrile_temps = []
for temp in temps_c:
    if temp >= 38.0:
        febrile_temps.append(temp)

print("Febrile patients (>= 38.0 C):", febrile_temps)
```

## To-Do 5.3 (`continue` Pattern): Skipping DHS Missing Codes (`999` / `None`)

A household survey exported respondent ages with DHS missing codes:
`raw_ages = [24, 999, 31, None, 28, 999, 37]`.

Use a `for` loop with `continue` to skip `None` and `999`, and compute:
- `valid_count` (`int`): count of valid ages
- `mean_age` (`float`): average of valid ages, rounded to **1 decimal place**

```python
raw_ages = [24, 999, 31, None, 28, 999, 37]

age_sum = 0
valid_count = 0
for age in raw_ages:
    if age is None or age == 999:
        continue
    age_sum += age
    valid_count += 1
mean_age = round(age_sum / valid_count, 1)

print(f"Valid ages: {valid_count} | Clean mean age: {mean_age} years")
```

```python
tutor.check("ch05_todo_03", (valid_count, mean_age))
```

## To-Do 5.4 (`break` + `enumerate()`): First Outbreak Threshold Breach

Daily dengue counts are `daily_cases = [18, 29, 44, 52, 67, 58]` and `alert_threshold = 50`.

Loop with `enumerate(daily_cases, start=1)` and stop immediately with `break` on the **first** day where `cases >= alert_threshold`.
Store:
- `breach_day` (`int`, 1-based day number)
- `breach_cases` (`int`, cases on that day)

```python
daily_cases = [18, 29, 44, 52, 67, 58]
alert_threshold = 50

breach_day = None
breach_cases = None
for day, cases in enumerate(daily_cases, start=1):
    if cases >= alert_threshold:
        breach_day = day
        breach_cases = cases
        break

print(f"First breach: Day {breach_day} ({breach_cases} cases)")
```

```python
tutor.check("ch05_todo_04", (breach_day, breach_cases))
```

## To-Do 5.5 (`while` Loop): Epidemic Doubling Simulation

Start with `cases = 15`, `bed_limit = 200`, and `days_elapsed = 0`.  
Write a `while` loop that doubles `cases` (`cases *= 2`) and increments `days_elapsed += 2` (2-day doubling time) **while `cases <= bed_limit`**.

```python
cases = 15
bed_limit = 200
days_elapsed = 0

while cases <= bed_limit:
    cases *= 2
    days_elapsed += 2

print(f"Bed limit exceeded on Day {days_elapsed} with {cases} cases")
```

```python
tutor.check("ch05_todo_05", (days_elapsed, cases))
```

## To-Do 5.6 (Nested Loops + `enumerate()`): Multi-District Clinic Totals

Three districts each report weekly admissions from three Upazila clinics:
`district_clinics = [[12, 18, 15], [22, 19, 25], [10, 14, 16]]`.

Use nested loops to compute:
- `district_totals` (`list[int]`): total cases for each district (`[45, 66, 40]`)
- `top_district` (`int`): 1-based district number with the highest total

```python
district_clinics = [
    [12, 18, 15],
    [22, 19, 25],
    [10, 14, 16],
]

district_totals = []
top_district = 1
best_total = -1

for d_num, clinics in enumerate(district_clinics, start=1):
    d_total = 0
    for count in clinics:
        d_total += count
    district_totals.append(d_total)
    if d_total > best_total:
        best_total = d_total
        top_district = d_num

print(f"District totals: {district_totals} | Top district: District {top_district}")
```

```python
tutor.check("ch05_todo_06", (district_totals, top_district))
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 5.3 (Denominator trap):** Dividing `age_sum / len(raw_ages)` divides by `7` instead of `valid_count = 4`, dragging the clean mean age down from `30.0` to `17.1` years!
- **To-Do 5.4 (Missing `break`):** Without `break`, the loop continues past Day 4 (`52`) and overwrites `breach_day` with Day 6 (`58`), misreporting the outbreak onset by 2 full days.
- **To-Do 5.6 (Inner accumulator reset):** Setting `d_total = 0` *inside* the outer loop ensures each district starts cleanly at `0`; initializing it outside produces cumulative running sums `[45, 111, 151]` instead of per-district totals.
