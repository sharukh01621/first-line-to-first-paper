## Exam Task 4.1 (30 Points): WHO BMI & Maternal Anemia Dual Classifier

An antenatal clinic record has `weight_kg = 46.0`, `height_m = 1.60`, and `hb_g_dl = 6.7`.

Compute:
1. `bmi`: `round(weight_kg / (height_m ** 2), 2)`
2. `bmi_category`: WHO adult category (`"Underweight"` `< 18.5`, `"Normal"` `< 25.0`, `"Overweight"` `< 30.0`, else `"Obese"`)
3. `anemia_severity`: guard `hb_g_dl is None or hb_g_dl == 999` -> `"Missing"`; otherwise `< 7.0` `"Severe"`, `< 10.0` `"Moderate"`, `< 11.0` `"Mild"`, else `"Normal"`
4. `high_risk_nutrition`: `bool` — `True` if `bmi_category == "Underweight"` AND `anemia_severity` is `"Severe"` or `"Moderate"`

```python
weight_kg = 46.0
height_m = 1.60
hb_g_dl = 6.7

bmi = round(weight_kg / (height_m ** 2), 2)

if bmi < 18.5:
    bmi_category = "Underweight"
elif bmi < 25.0:
    bmi_category = "Normal"
elif bmi < 30.0:
    bmi_category = "Overweight"
else:
    bmi_category = "Obese"

if hb_g_dl is None or hb_g_dl == 999:
    anemia_severity = "Missing"
elif hb_g_dl < 7.0:
    anemia_severity = "Severe"
elif hb_g_dl < 10.0:
    anemia_severity = "Moderate"
elif hb_g_dl < 11.0:
    anemia_severity = "Mild"
else:
    anemia_severity = "Normal"

high_risk_nutrition = (
    bmi_category == "Underweight"
    and (anemia_severity == "Severe" or anemia_severity == "Moderate")
)

print(f"BMI: {bmi} ({bmi_category}) | Anemia: {anemia_severity} | High Risk: {high_risk_nutrition}")
```

## Exam Task 4.2 (35 Points): Upazila Clinic Emergency Dengue Triage

A febrile child arrives at an Upazila Health Complex with:
`temp_c = 39.4`, `platelets_k = 42` (thousands/µL), `has_bleeding = False`, `age_years = 9`.

Compute:
1. `critical_thrombocytopenia`: `platelets_k < 50` (`bool`)
2. `pediatric_high_fever`: `age_years < 12 and temp_c >= 39.0` (`bool`)
3. `triage_priority`:
   - `"Urgent"` if `has_bleeding or critical_thrombocytopenia or temp_c >= 40.0`
   - `"Moderate"` if `(50 <= platelets_k < 100) or (temp_c >= 38.5)`
   - `"Routine"` otherwise
4. `admit_inpatient`: `True` if `triage_priority == "Urgent"` OR (`triage_priority == "Moderate"` and `age_years < 12`)

```python
temp_c = 39.4
platelets_k = 42
has_bleeding = False
age_years = 9

critical_thrombocytopenia = platelets_k < 50
pediatric_high_fever = (age_years < 12) and (temp_c >= 39.0)

if has_bleeding or critical_thrombocytopenia or temp_c >= 40.0:
    triage_priority = "Urgent"
elif (50 <= platelets_k < 100) or (temp_c >= 38.5):
    triage_priority = "Moderate"
else:
    triage_priority = "Routine"

admit_inpatient = (
    triage_priority == "Urgent"
    or (triage_priority == "Moderate" and age_years < 12)
)

print(f"Triage: {triage_priority} | Admit: {admit_inpatient}")
```

## Exam Task 4.3 (35 Points): Safe Outbreak Surveillance Guard Pipeline

Two rural wards submit weekly cholera surveillance records:
- **Ward A (valid):** `cases_a = 28`, `pop_a = 40000`
- **Ward B (corrupted):** `cases_b = 999` (missing sentinel!), `pop_b = 0` (zero census!)

For **each** ward, guard against invalid records (`pop <= 0`, `cases is None`, `cases == 999`, or `cases < 0`) BEFORE dividing:
- If valid: compute `rate = round((cases / pop) * 100000, 2)` and assign `alert`: `"Red"` (`>= 50.0`), `"Amber"` (`>= 20.0`), else `"Green"`.
- If invalid: set `rate = None` and `alert = "Data Error"`.

Produce `rate_a`, `alert_a`, `rate_b`, and `alert_b` without raising `ZeroDivisionError`.

```python
cases_a, pop_a = 28, 40000
cases_b, pop_b = 999, 0

if pop_a > 0 and cases_a is not None and cases_a != 999 and cases_a >= 0:
    rate_a = round((cases_a / pop_a) * 100000, 2)
    if rate_a >= 50.0:
        alert_a = "Red"
    elif rate_a >= 20.0:
        alert_a = "Amber"
    else:
        alert_a = "Green"
else:
    rate_a = None
    alert_a = "Data Error"

if pop_b > 0 and cases_b is not None and cases_b != 999 and cases_b >= 0:
    rate_b = round((cases_b / pop_b) * 100000, 2)
    if rate_b >= 50.0:
        alert_b = "Red"
    elif rate_b >= 20.0:
        alert_b = "Amber"
    else:
        alert_b = "Green"
else:
    rate_b = None
    alert_b = "Data Error"

print(f"Ward A: {rate_a} ({alert_a}) | Ward B: {rate_b} ({alert_b})")
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 4.1 (Operator Precedence Trap):** Writing `bmi_category == "Underweight" and anemia_severity == "Severe" or anemia_severity == "Moderate"` **without parentheses** around the `or` clause evaluates `(Underweight and Severe) or Moderate` because `and` binds tighter than `or`—flagging a normal-BMI mother with Moderate anemia as `high_risk_nutrition = True`! Always parenthesize mixed `and` / `or` logic.
- **Task 4.2 (Triage Ordering):** Placing `Moderate` (`temp_c >= 38.5`) *above* `Urgent` (`platelets_k < 50`) would dangerously downgrade a child with critical thrombocytopenia (`42k`) and `39.4°C` fever from `"Urgent"` to `"Moderate"`!
