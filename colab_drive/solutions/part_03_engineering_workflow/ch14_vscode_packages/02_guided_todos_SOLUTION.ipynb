## To-Do 14.1 (Warm-Up): Parse a `requirements.txt` File

Write `parse_requirements_file(req_path)` that reads a UTF-8 requirements file:
- Ignore blank lines and lines starting with `#`.
- Split each dependency line on `==`, `>=`, `<=`, or `~=` (or accept a bare package name with no specifier).
- Return a dictionary mapping **lowercased package name** $\rightarrow$ **full version specifier** (or `""` for bare names), e.g. `{"pandas": "==2.2.0", "numpy": ">=1.26", "scipy": ""}`.

```python
def parse_requirements_file(req_path):
    """Parse requirements.txt into dict of pkg_lower -> spec_or_empty."""
    reqs = {}
    with open(req_path, "r", encoding="utf-8") as f:
        for raw in f:
            line = raw.strip()
            if not line or line.startswith("#"):
                continue
            for op in ("==", ">=", "<=", "~="):
                if op in line:
                    name, version = line.split(op, 1)
                    reqs[name.strip().lower()] = op + version.strip()
                    break
            else:
                reqs[line.lower()] = ""
    return reqs
```

## To-Do 14.2 (Reproducibility Check): Find Unpinned Packages

Write `find_unpinned_packages(req_dict)` that takes a `pkg -> spec` dictionary and **returns** a **sorted list** of package names whose specifier does **not** start with `"=="`.

```python
def find_unpinned_packages(req_dict):
    """Return sorted list of package names not pinned with '=='."""
    return sorted(
        pkg for pkg, spec in req_dict.items()
        if not spec.startswith("==")
    )
```

## To-Do 14.4 (Package Hygiene): Validate Standard `src/` Layout

Write `validate_src_package_layout(root_dir, pkg_name)` that checks inside `root_dir` for:
- `pyproject.toml` (file)
- `src/<pkg_name>/__init__.py` (file)
- `tests/` (directory)

Return `{"has_pyproject": bool, "has_init": bool, "has_tests": bool, "is_valid": bool}` where `is_valid` is `True` only when all three checks pass.

```python
from pathlib import Path

def validate_src_package_layout(root_dir, pkg_name):
    """Check pyproject.toml, src/<pkg>/__init__.py, and tests/ dir."""
    root = Path(root_dir)
    has_pyproject = (root / "pyproject.toml").is_file()
    has_init = (root / "src" / pkg_name / "__init__.py").is_file()
    has_tests = (root / "tests").is_dir()
    return {
        "has_pyproject": has_pyproject,
        "has_init": has_init,
        "has_tests": has_tests,
        "is_valid": has_pyproject and has_init and has_tests,
    }
```

## To-Do 14.5 (Import Naming): Normalize Python Package Name

Write `normalize_package_name(raw_name)` that converts a raw project title into a clean Python import name:
1. Strip outer whitespace and lowercase.
2. Replace hyphens (`-`) and spaces (` `) with underscores (`_`).
3. Collapse repeated underscores (`__`, `___`) into a single `_`.
4. Strip any leading or trailing `_` and return the result.

```python
import re

def normalize_package_name(raw_name):
    """Normalize raw project name into snake_case package name."""
    cleaned = raw_name.strip().lower()
    cleaned = cleaned.replace("-", "_").replace(" ", "_")
    cleaned = re.sub(r"_+", "_", cleaned)
    return cleaned.strip("_")
```

## To-Do 14.6 (Config & Secrets): Load a `.env` File

Write `load_env_file(env_path)` that reads a UTF-8 `.env` file:
- Ignore blank lines and lines starting with `#`.
- Parse `KEY=VALUE` lines using `split("=", 1)` so values containing `=` (such as query URLs) stay intact.
- Strip whitespace around keys and values, remove matching surrounding single or double quotes (`'...'` or `"..."`) from values, and return the resulting dictionary.

```python
def load_env_file(env_path):
    """Parse KEY=VALUE pairs from .env ignoring comments and blanks."""
    env = {}
    with open(env_path, "r", encoding="utf-8") as f:
        for raw in f:
            line = raw.strip()
            if not line or line.startswith("#"):
                continue
            key, val = line.split("=", 1)
            key, val = key.strip(), val.strip()
            if len(val) >= 2 and val[0] == val[-1]:
                if val[0] in ('"', "'"):
                    val = val[1:-1]
            env[key] = val
    return env
```
