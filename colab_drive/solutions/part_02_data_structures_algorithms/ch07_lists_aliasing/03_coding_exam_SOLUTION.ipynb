## Exam Task 7.1 (30 Points): Clean & Sort Systolic Blood Pressure

Write a pure function `clean_and_sort_sbp(readings)` that processes a list of raw systolic blood pressure (mmHg) readings:
1. Skip all `None` entries, sentinel `999` values, and non-positive values (`<= 0`).
2. Return a **new list** of the valid readings sorted in **ascending** order.
3. Do **not** mutate the caller's original `readings` list.

```python
def clean_and_sort_sbp(readings):
    """Return sorted ascending valid SBP list (skip None, 999, <= 0)."""
    valid = [
        val for val in readings
        if val is not None and val != 999 and val > 0
    ]
    return sorted(valid)
```

## Exam Task 7.2 (35 Points): Rolling 3-Day Outbreak Totals

Write `rolling_3day_totals(daily_cases)` that computes sliding 3-day cumulative case totals across a daily outbreak series:
- For each consecutive 3-day window `daily_cases[i : i + 3]`, compute its sum and append it to the result list.
- If `len(daily_cases) < 3`, return an empty list `[]`.
- Example: `[10, 15, 20, 25, 30]` $\rightarrow$ `[45, 60, 75]`.
- Do **not** mutate `daily_cases`.

```python
def rolling_3day_totals(daily_cases):
    """Return sliding 3-day window sums, or [] if fewer than 3 days."""
    if len(daily_cases) < 3:
        return []
    return [
        sum(daily_cases[i : i + 3])
        for i in range(len(daily_cases) - 2)
    ]
```

## Exam Task 7.3 (35 Points): Flag & Adjust Medication Doses

Write `flag_and_adjust_doses(doses_mg, cap_mg)` to enforce a safe upper dosing ceiling on a batch of prescribed doses:
1. Create a **new** list `adjusted_list` where any dose strictly greater than `cap_mg` (`dose > cap_mg`) is capped down to `cap_mg`, while doses `<= cap_mg` stay unchanged.
2. Count `capped_count` (`int`): how many doses were strictly greater than `cap_mg`.
3. Return the 2-tuple `(adjusted_list, capped_count)` without mutating the caller's original `doses_mg` list.

```python
def flag_and_adjust_doses(doses_mg, cap_mg):
    """Cap doses > cap_mg without mutating input; return (list, count)."""
    adjusted = []
    capped_count = 0
    for dose in doses_mg:
        if dose > cap_mg:
            adjusted.append(cap_mg)
            capped_count += 1
        else:
            adjusted.append(dose)
    return (adjusted, capped_count)
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 7.1 (`None` Filter Order & `sorted()` vs. `.sort()`):** Always test `val is not None` *before* `val > 0`, otherwise Python raises `TypeError` on `None > 0`. Using `sorted(valid)` returns a new ascending list instead of `None`.
- **Task 7.2 (Sliding Window Bounds):** For a 3-day window over $N$ days, valid start indices run from `0` to `N - 3` inclusive—which is `range(len(daily_cases) - 2)`. Stopping at `N - 3` drops the final window, while looping to `N` sums incomplete 1-day and 2-day tails.
- **Task 7.3 (Non-Mutating Audit Trail & Strict `>`):** Overwriting `doses_mg[i] = cap_mg` destroys the clinician's original prescription log! Building a fresh `adjusted` list and testing strictly `dose > cap_mg` preserves the audit trail and avoids wrongly flagging doses equal to `cap_mg`.
