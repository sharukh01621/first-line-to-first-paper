## Exam Task 22.1 (30 Points): `compare_ols_ridge_lasso(X_train, y_train, X_test, y_test, ridge_alpha=1.0, lasso_alpha=0.1)`

Benchmark three leak-free standardized regression pipelines (`OLS`, `Ridge`, `Lasso`) without mutating inputs:
1. Wrap `LinearRegression()`, `Ridge(alpha=ridge_alpha)`, and `Lasso(alpha=lasso_alpha, max_iter=5000)` in `Pipeline([("scaler", StandardScaler()), ("model", est)])`.
2. Fit each pipeline on `(X_train, y_train)` and predict on `X_test`.
3. Compute `rmse`, `mae`, and `r2` rounded to **4 decimals**, plus `nonzero_coefs` (`int(np.sum(np.abs(coef) > 1e-6))`).
4. Return a `pd.DataFrame` with exact columns `["model", "rmse", "mae", "r2", "nonzero_coefs"]` and rows ordered `["OLS", "Ridge", "Lasso"]`.

```python
def compare_ols_ridge_lasso(
    X_train, y_train, X_test, y_test,
    ridge_alpha=1.0, lasso_alpha=0.1,
):
    """Benchmark standardized OLS, Ridge, and Lasso on a holdout set."""
    specs = [
        ("OLS", LinearRegression()),
        ("Ridge", Ridge(alpha=ridge_alpha)),
        ("Lasso", Lasso(alpha=lasso_alpha, max_iter=5000)),
    ]
    rows = []
    for name, est in specs:
        pipe = Pipeline([
            ("scaler", StandardScaler()),
            ("model", est),
        ])
        pipe.fit(X_train, y_train)
        pred = pipe.predict(X_test)
        coefs = pipe.named_steps["model"].coef_
        rows.append({
            "model": name,
            "rmse": round(
                float(root_mean_squared_error(y_test, pred)), 4
            ),
            "mae": round(
                float(mean_absolute_error(y_test, pred)), 4
            ),
            "r2": round(float(r2_score(y_test, pred)), 4),
            "nonzero_coefs": int(np.sum(np.abs(coefs) > 1e-6)),
        })
    return pd.DataFrame(rows)
```

## Exam Task 22.2 (35 Points): `tune_ridge_alpha_cv(X, y, alphas, cv=5)`

Select the optimal Ridge penalty `alpha` via leak-free $K$-fold cross-validation:
1. Instantiate `kf = KFold(n_splits=cv, shuffle=True, random_state=42)`.
2. For each `a` in `alphas`, evaluate `Pipeline([("scaler", StandardScaler()), ("ridge", Ridge(alpha=float(a)))])` via `cross_val_score(..., cv=kf, scoring="neg_root_mean_squared_error")`.
3. Negate the scores (`rmses = -scores`) and build `cv_table` with columns `["alpha", "mean_rmse", "std_rmse"]` (`mean_rmse` and `std_rmse` rounded to **4 decimals**).
4. Return `{"best_alpha": float, "best_rmse": round4, "cv_table": DataFrame}` choosing the `alpha` with minimum `mean_rmse`.

```python
def tune_ridge_alpha_cv(X, y, alphas, cv=5):
    """Tune Ridge alpha via leak-free KFold CV on positive RMSE."""
    kf = KFold(n_splits=cv, shuffle=True, random_state=42)
    rows = []
    for a in alphas:
        pipe = Pipeline([
            ("scaler", StandardScaler()),
            ("ridge", Ridge(alpha=float(a))),
        ])
        rmses = -cross_val_score(
            pipe, X, y, cv=kf,
            scoring="neg_root_mean_squared_error",
        )
        rows.append({
            "alpha": float(a),
            "mean_rmse": round(float(rmses.mean()), 4),
            "std_rmse": round(float(rmses.std()), 4),
        })
    cv_table = pd.DataFrame(rows)
    best_idx = int(cv_table["mean_rmse"].idxmin())
    return {
        "best_alpha": float(cv_table.loc[best_idx, "alpha"]),
        "best_rmse": float(cv_table.loc[best_idx, "mean_rmse"]),
        "cv_table": cv_table,
    }
```

## Exam Task 22.3 (35 Points): `predict_child_haz_pipeline(mch_df, seed=42)`

Build an end-to-end Child Height-for-Age Z-score (`child_haz`) Ridge prediction pipeline without mutating `mch_df`:
1. Copy `mch_df`, standardize `residence` (`str.strip().str.title()`), replace `anc_visits == 999` with `np.nan`, and add `is_urban = (work["residence"] == "Urban").astype(int)`.
2. Drop rows with NaNs on `["anc_visits", "maternal_hb_g_dl", "maternal_age_years", "is_urban", "child_haz"]`.
3. Split features (`["anc_visits", "maternal_hb_g_dl", "maternal_age_years", "is_urban"]`) and target (`child_haz`) 80/20 via `train_test_split(..., test_size=0.2, random_state=seed)`.
4. Fit `Pipeline([("scaler", StandardScaler()), ("ridge", Ridge(alpha=1.0))])` on the training split and evaluate on the test split.
5. Return `{"n_clean": int, "test_rmse": round4, "test_r2": round4, "coefs": dict_feature_to_round4}`.

```python
def predict_child_haz_pipeline(mch_df, seed=42):
    """Clean MCH survey data and fit a standardized Ridge HAZ model."""
    work = mch_df.copy()
    work["residence"] = (
        work["residence"].astype(str).str.strip().str.title()
    )
    work["anc_visits"] = work["anc_visits"].replace(999, np.nan)
    work["is_urban"] = (work["residence"] == "Urban").astype(int)

    feats = [
        "anc_visits", "maternal_hb_g_dl",
        "maternal_age_years", "is_urban",
    ]
    clean = work[feats + ["child_haz"]].dropna()
    X_tr, X_te, y_tr, y_te = train_test_split(
        clean[feats], clean["child_haz"],
        test_size=0.2, random_state=seed,
    )
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=1.0)),
    ])
    pipe.fit(X_tr, y_tr)
    y_pred = pipe.predict(X_te)
    coefs = {
        f: round(float(c), 4)
        for f, c in zip(feats, pipe.named_steps["ridge"].coef_)
    }
    return {
        "n_clean": int(len(clean)),
        "test_rmse": round(
            float(root_mean_squared_error(y_te, y_pred)), 4
        ),
        "test_r2": round(float(r2_score(y_te, y_pred)), 4),
        "coefs": coefs,
    }
```

## Chapter 22 Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 22.1 (Forgetting `StandardScaler` or `max_iter=5000`):** Without `StandardScaler` inside each `Pipeline`, `Ridge` and `Lasso` penalize coefficients in raw physical units rather than standard deviations, skewing both test error and sparsity counts. Passing `max_iter=5000` also prevents coordinate-descent convergence warnings on collinear features.
- **Task 22.2 (Sign Flip on `neg_root_mean_squared_error`):** Scikit-Learn follows a "higher is better" convention for all scorers, so `cross_val_score(..., scoring='neg_root_mean_squared_error')` returns **negative** numbers. Always negate (`rmses = -scores`) before computing `mean_rmse` and choosing `idxmin()`.
- **Task 22.3 (Sentinel `999` & DataFrame Mutation):** Leaving `anc_visits == 999` uncleaned inflates its training mean/SD dramatically and biases both scaling and Ridge weights. Always work on `mch_df.copy()`, never use `inplace=True` (Pandas 3.x), and keep `StandardScaler` inside the `Pipeline` fitted strictly on `X_tr`.
