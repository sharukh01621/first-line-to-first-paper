## Section 7.3.1 — Zero-Based Indexing, Negative Indexing, and Slicing

```python
weekly_cases = [12, 18, 25, 31, 40, 28]

print("First week (index 0):", weekly_cases[0])
print("Last week (index -1):", weekly_cases[-1])
print("Weeks 2 to 4 [1:4]:  ", weekly_cases[1:4])
print("Last 3 weeks [-3:]:  ", weekly_cases[-3:])
print("Every 2nd week [::2]:", weekly_cases[::2])
print("Reversed copy [::-1]:", weekly_cases[::-1])
```

## Section 7.3.2 — In-Place Mutation Methods vs. Non-Mutating `sorted()`

```python
sbp_log = [124, 138, 119]

# In-place mutation methods modify sbp_log directly and return None
ret_val = sbp_log.append(145)
print("After append(145):", sbp_log, "| return value:", ret_val)

sbp_log.extend([131, 128])
sbp_log.insert(0, 120)
popped = sbp_log.pop()
print("After extend/insert/pop:", sbp_log, "| popped:", popped)

# sorted() returns a NEW sorted list; original chronological order stays!
ranked = sorted(sbp_log, reverse=True)
print("Ranked copy:     ", ranked)
print("Original intact: ", sbp_log)
```

## Section 7.3.3 — The Aliasing Trap: `b = a` vs. `.copy()` vs. `copy.deepcopy()`

```python
import copy

# 1. Direct assignment creates an ALIAS (same list in memory!)
raw_hb = [10.4, 11.8, 12.1]
alias_hb = raw_hb
alias_hb[0] = 999
print("Alias -> raw_hb changed:", raw_hb, "same?", alias_hb is raw_hb)

# 2. Shallow .copy() isolates flat 1D lists
clean_hb = [10.4, 11.8, 12.1]
safe_1d = clean_hb.copy()
safe_1d[0] = 9.9
print("Flat .copy() -> intact:", clean_hb, "| copy:", safe_1d)

# 3. Nested 2D ward matrix: shallow copy SHARES inner rows!
wards = [[12, 14], [18, 20]]
shallow_wards = wards.copy()
deep_wards = copy.deepcopy(wards)

shallow_wards[0][0] = 999
print("After shallow edit -> wards corrupted:", wards)
print("Deep copy unaffected:                 ", deep_wards)
```

## Section 7.3.4 — List Comprehensions: Clean Filtering & Transformation

```python
raw_readings = [10.4, None, 11.8, 999, 9.6, 12.3]
heights_cm = [168, 156, 172.4, 160]

# Filter sentinels into a brand-new list (raw_readings untouched)
valid_hb = [v for v in raw_readings if v is not None and v != 999]

# Transform units cleanly into a new list
heights_m = [round(h / 100, 2) for h in heights_cm]

# Filter + transform in one readable expression
anemic_gaps = [round(11.0 - v, 2) for v in valid_hb if v < 11.0]

print("Valid Hb:     ", valid_hb)
print("Heights (m):  ", heights_m)
print("Anemic gaps:  ", anemic_gaps)
print("Raw unchanged:", raw_readings)
```

## Section 7.4 — Worked Problem-Structuring Blueprint: Non-Mutating Outbreak Peak Analyzer

```python
def analyze_surveillance_series(raw_weekly_cases, window=3, top_k=2):
    """Clean sentinels, slice recent window, and find top-k peaks safely."""
    cleaned = [
        c for c in raw_weekly_cases
        if c is not None and c != 999 and c >= 0
    ]
    recent = [] if window <= 0 else cleaned[-window:]
    peaks = [] if top_k <= 0 else sorted(cleaned, reverse=True)[:top_k]
    return (cleaned, recent, peaks)

dhaka_raw = [14, None, 42, 999, 19, 68, -3, 35, 51]
cleaned, recent_3, top_2 = analyze_surveillance_series(dhaka_raw)

print("Cleaned series:", cleaned)
print("Last 3 weeks:  ", recent_3)
print("Top 2 peaks:   ", top_2)
print("Raw untouched: ", dhaka_raw)
```
