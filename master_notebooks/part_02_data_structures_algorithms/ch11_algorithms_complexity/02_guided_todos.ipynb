## To-Do 11.1 (Linear Search — $O(n)$): Find a Patient ID

Write `linear_search_patient(patient_ids, target_id)` that scans `patient_ids` left-to-right and **returns** the 0-based `int` index of the first occurrence of `target_id`, or `-1` if `target_id` is not present.

```python
def linear_search_patient(patient_ids, target_id):
    """Return 0-based index of target_id in patient_ids, or -1."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    for idx, pid in enumerate(patient_ids):
        if pid == target_id:
            return idx
    return -1
    ### END SOLUTION
```

## To-Do 11.2 (Binary Search — $O(\log n)$): Iterative Sorted Lookup

Write `binary_search_sorted(sorted_ids, target_id)` using an iterative `while low <= high:` loop that halves the search range each step and **returns** the 0-based index of `target_id`, or `-1` if not found.

*Note:* The verifier checks both boundary cases **and** comparison count to confirm true $O(\log n)$ behavior!

```python
def binary_search_sorted(sorted_ids, target_id):
    """Iterative O(log n) binary search returning index or -1."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    low = 0
    high = len(sorted_ids) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_ids[mid] == target_id:
            return mid
        elif sorted_ids[mid] < target_id:
            low = mid + 1
        else:
            high = mid - 1
    return -1
    ### END SOLUTION
```

## To-Do 11.3 (Hash-Set Speedup — $O(n)$): Fast Duplicate ID Finder

Write `find_duplicate_ids_fast(patient_ids)` using Python sets (`seen` and `duplicates`) so duplicate detection runs in $O(n)$ time:
- Return a **sorted list** of unique duplicate patient IDs (each duplicate listed once).
- Do **not** mutate `patient_ids`.
- The verifier benchmarks 8,000 IDs so quadratic $O(n^2)$ `.count()` or list-membership scans are caught!

```python
def find_duplicate_ids_fast(patient_ids):
    """Return sorted list of duplicate patient IDs in O(n) time."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    seen = set()
    duplicates = set()
    for pid in patient_ids:
        if pid in seen:
            duplicates.add(pid)
        else:
            seen.add(pid)
    return sorted(duplicates)
    ### END SOLUTION
```

```python
tutor.check("ch11_todo_03", find_duplicate_ids_fast)
```

## To-Do 11.4 (Multi-Key Sorting — $O(n \log n)$): Non-Mutating Triage Queue Sorter

Write `sort_triage_queue(patients)` accepting a list of patient dicts (`{"patient_id": ..., "urgency": ..., "temp_c": ...}`):
- Return a **new** sorted list ordered by clinical urgency (`"Urgent"` first, then `"Moderate"`, then `"Routine"`), and within the same urgency by **descending** `temp_c` (highest fever first).
- Do **not** mutate the input list `patients` (use `sorted()`, not `.sort()`).

```python
def sort_triage_queue(patients):
    """Return new list sorted by urgency rank then descending temp_c."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    urgency_rank = {"Urgent": 0, "Moderate": 1, "Routine": 2}
    return sorted(
        patients,
        key=lambda p: (urgency_rank[p["urgency"]], -p["temp_c"]),
    )
    ### END SOLUTION
```

## To-Do 11.5 (Linear Recursion): Outbreak Generation Projection

Write a **recursive** function `recursive_outbreak_cases(initial_cases, r_factor, generations)`:
- **Base case:** When `generations == 0`, return `round(initial_cases, 2)`.
- **Recursive step:** Compute the previous generation recursively (`generations - 1`), multiply by `r_factor`, and return `round(..., 2)`.

```python
def recursive_outbreak_cases(initial_cases, r_factor, generations):
    """Recursively project outbreak cases after N generations."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    if generations == 0:
        return round(initial_cases, 2)
    prev = recursive_outbreak_cases(
        initial_cases, r_factor, generations - 1
    )
    return round(prev * r_factor, 2)
    ### END SOLUTION
```

## To-Do 11.6 (Tree Recursion): Contact-Tracing Chain Counter

Write `count_contact_chain(contact_tree, root_id)` where `contact_tree` maps each person ID to a list of direct contact IDs (an acyclic transmission tree):
- Recursively count and **return** the total number of reachable individuals **including `root_id` itself**.
- Use `contact_tree.get(root_id, [])` so leaf contacts omitted from the dictionary never raise `KeyError`.

```python
def count_contact_chain(contact_tree, root_id):
    """Recursively count reachable people in contact_tree incl. root."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    total = 1
    for contact in contact_tree.get(root_id, []):
        total += count_contact_chain(contact_tree, contact)
    return total
    ### END SOLUTION
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 11.2 (`while low <= high`):** Using `while low < high:` skips the final single-element check when `low == high`, causing misses on the first or last patient ID.
- **To-Do 11.3 (`set` vs. `list`):** Checking `pid in seen_list` or calling `patient_ids.count(pid)` inside a `for` loop silently turns an $O(n)$ task into $O(n^2)$—which stalls on national registries.
- **To-Do 11.4 (Alphabetical Trap & Mutation):** Sorting by `p["urgency"]` directly puts `"Moderate"` before `"Urgent"` alphabetically! Mapping urgency to integer ranks and using `sorted()` instead of `.sort()` solves both bugs cleanly.
- **To-Do 11.6 (`KeyError` on Leaf Nodes):** Contact trees rarely store empty lists for terminal contacts. Calling `contact_tree.get(root_id, [])` provides a built-in base case without crashing.
