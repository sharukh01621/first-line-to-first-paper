## Section 5.3.1 — `for` Loops & `range()` (Half-Open Interval `[start, stop)`)

```python
# range(start, stop, step) stops BEFORE 'stop' (half-open interval)
for day in range(1, 6):
    print(f"Day {day}: Surveillance report submitted")

print("Days counted:", list(range(1, 6)))
print("Even clinic hours:", list(range(8, 17, 2)))
```

## Section 5.3.2 — The 4 Universal Loop Patterns (Accumulator, Counter, Filter, Running Max)

```python
hb_readings = [10.4, 12.1, 9.6, 11.8, 10.9]

total_hb = 0.0           # Pattern 1: Accumulator
anemic_count = 0         # Pattern 2: Counter
anemic_cases = []        # Pattern 3: Filter
max_hb = hb_readings[0]  # Pattern 4: Running Max

for hb in hb_readings:
    total_hb += hb
    if hb < 11.0:
        anemic_count += 1
        anemic_cases.append(hb)
    if hb > max_hb:
        max_hb = hb

mean_hb = round(total_hb / len(hb_readings), 2)
print("Mean Hb:", mean_hb, "| Anemic count:", anemic_count)
print("Anemic list:", anemic_cases, "| Highest Hb:", max_hb)
```

## Section 5.3.3 — `continue` & `break` (DHS Missing Codes & Early Outbreak Alert)

```python
# Part A: continue skips DHS missing sentinels (999 or None)
raw_sbp = [122, 999, 148, None, 134, 999]
valid_total = 0
valid_count = 0

for sbp in raw_sbp:
    if sbp is None or sbp == 999:
        continue
    valid_total += sbp
    valid_count += 1

print("Clean SBP mean:", round(valid_total / valid_count, 1))

# Part B: break halts scan on FIRST outbreak threshold breach (>= 50)
daily_cases = [14, 28, 53, 61, 40]
first_alert_day = None

for day_num, cases in enumerate(daily_cases, start=1):
    if cases >= 50:
        first_alert_day = day_num
        break

print("First outbreak alert on Day:", first_alert_day)
```

## Section 5.3.4 — `enumerate()` for Peak Dengue Week Identification

```python
# Weekly dengue cases across 6 monsoon weeks
weekly_cases = [32, 48, 79, 115, 94, 61]
peak_week = 1
peak_cases = weekly_cases[0]

for week, cases in enumerate(weekly_cases, start=1):
    if cases > peak_cases:
        peak_cases = cases
        peak_week = week

print(f"Peak dengue week: Week {peak_week} ({peak_cases} cases)")
```

## Section 5.3.5 — `while` Loops: Epidemic Doubling & Hospital Bed Capacity

```python
# Simulate cases doubling every 3 days until exceeding 200 beds
active_cases = 25
bed_capacity = 200
days = 0

while active_cases <= bed_capacity:
    active_cases *= 2
    days += 3

print(f"Capacity exceeded on Day {days} with {active_cases} cases")
```

## Section 5.3.6 — Nested Loops: Multi-Ward Surveillance Summary

```python
# Each inner list holds 3 daily dengue admissions for one ward
ward_admissions = [
    [6, 9, 11],    # Ward 1 -> sum 26
    [14, 18, 12],  # Ward 2 -> sum 44
]
grand_total = 0

for ward_num, daily_counts in enumerate(ward_admissions, start=1):
    ward_total = 0
    for count in daily_counts:
        ward_total += count
    grand_total += ward_total
    print(f"Ward {ward_num} 3-day total: {ward_total}")

print("Hospital grand total:", grand_total)
```

## Section 5.4 — Worked Problem-Structuring Blueprint: Single-Pass Maternal Hb Cleaner

```python
# Step 1: Raw maternal hemoglobin readings with DHS missing codes
raw_hb = [10.2, 999, 11.6, None, 9.4, 12.3, 10.8]

# Step 2: Initialize accumulators, counters, and safe running min
valid_sum = 0.0
valid_count = 0
anemic_count = 0
min_hb = None

# Step 3: Single-pass cleaning + summary loop
for hb in raw_hb:
    if hb is None or hb == 999:
        continue
    valid_sum += hb
    valid_count += 1
    if hb < 11.0:
        anemic_count += 1
    if min_hb is None or hb < min_hb:
        min_hb = hb

# Step 4: Guard against ZeroDivisionError before computing mean
if valid_count > 0:
    mean_hb = round(valid_sum / valid_count, 2)
else:
    mean_hb = None

print(f"Valid: {valid_count} | Mean Hb: {mean_hb} g/dL")
print(f"Anemic: {anemic_count} | Most severe (Min Hb): {min_hb} g/dL")
```
