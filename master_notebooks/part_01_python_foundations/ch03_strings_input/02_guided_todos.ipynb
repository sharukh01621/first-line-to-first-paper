## To-Do 3.1 (Warm-Up): Cleaning Messy Clinic District Names

A clinic intake sheet recorded a district name as `raw_district = "   sylhet_sadar   "`.

1. Create `clean_district` by stripping surrounding spaces, replacing `"_"` with a space `" "`, and converting to Title Case (`"Sylhet Sadar"`).
2. Compute `char_count = len(clean_district)` (`int`).

```python
raw_district = "   sylhet_sadar   "

# STUB: clean_district = None
# STUB: char_count = None
### BEGIN SOLUTION
clean_district = raw_district.strip().replace("_", " ").title()
char_count = len(clean_district)
### END SOLUTION

print(f"District: {clean_district!r} | Characters: {char_count}")
```

```python
tutor.check("ch03_todo_01", (clean_district, char_count))
```

## To-Do 3.2 (Indexing & Slicing): Structured Patient IDs

Given `patient_id = "BD-DHK-2026-0482"`, extract four fields using string slicing:
- `country_code`: `"BD"` (`str`)
- `district_code`: `"DHK"` (`str`)
- `cohort_year`: `2026` (`int`)
- `serial_str`: `"0482"` (`str`, using negative indexing `[-4:]` so the leading zero is preserved)

```python
patient_id = "BD-DHK-2026-0482"

# STUB: country_code = None
# STUB: district_code = None
# STUB: cohort_year = None
# STUB: serial_str = None
### BEGIN SOLUTION
country_code = patient_id[:2]
district_code = patient_id[3:6]
cohort_year = int(patient_id[7:11])
serial_str = patient_id[-4:]
### END SOLUTION

print(country_code, district_code, cohort_year, serial_str)
```

```python
tutor.check("ch03_todo_02", (country_code, district_code, cohort_year, serial_str))
```

## To-Do 3.3 (Splitting Delimited Text): Pipe-Delimited Lab Record

A hospital lab exported a pipe-delimited record:
`raw_record = "  PAT_109 | 38.4 C | dengue_ns1_pos  "`

Split `raw_record` on `"|"` and compute:
- `lab_id`: `"PAT_109"` (`str`)
- `temp_c`: `38.4` (`float`)
- `is_dengue_pos`: `True` (`bool`, checking whether `"pos"` appears in the third field)

```python
raw_record = "  PAT_109 | 38.4 C | dengue_ns1_pos  "

# STUB: lab_id = None
# STUB: temp_c = None
# STUB: is_dengue_pos = None
### BEGIN SOLUTION
parts = raw_record.split("|")
lab_id = parts[0].strip()
temp_c = float(parts[1].replace("C", "").strip())
is_dengue_pos = "pos" in parts[2].strip().lower()
### END SOLUTION

print(f"Lab ID: {lab_id} | Temp: {temp_c} C | Dengue Pos: {is_dengue_pos}")
```

```python
tutor.check("ch03_todo_03", (lab_id, temp_c, is_dengue_pos))
```

## To-Do 3.4 (Immutability & Rebuild): Standardizing Facility Codes

Because Python strings are **immutable**, you cannot edit characters in place.  
Given `raw_code = "uhc_0042"`:
1. Build `fixed_code` by uppercasing and replacing `"_"` with `"-"` (`"UHC-0042"`).
2. Build `replaced_facility` by swapping the 3-letter prefix to `"DHK"` using slicing + concatenation (`"DHK-0042"`).

```python
raw_code = "uhc_0042"

# STUB: fixed_code = None
# STUB: replaced_facility = None
### BEGIN SOLUTION
fixed_code = raw_code.upper().replace("_", "-")
replaced_facility = "DHK" + fixed_code[3:]
### END SOLUTION

print(f"Original: {raw_code} | Fixed: {fixed_code} | Reassigned: {replaced_facility}")
```

```python
tutor.check("ch03_todo_04", (fixed_code, replaced_facility))
```

## To-Do 3.5 (Publication Formatting): Epidemiology Summary f-String

Given `district = "Rajshahi"`, `screened = 12450`, `positive = 386`, `prevalence = positive / screened`, and `mean_hb = 11.246`, create `report_line` using a single f-string that produces **exactly**:

`"Rajshahi: n=12,450 | Pos=386 (3.10%) | Mean Hb=11.2 g/dL"`

```python
district = "Rajshahi"
screened = 12450
positive = 386
prevalence = positive / screened
mean_hb = 11.246

# STUB: report_line = None
### BEGIN SOLUTION
report_line = (
    f"{district}: n={screened:,} | Pos={positive} "
    f"({prevalence:.2%}) | Mean Hb={mean_hb:.1f} g/dL"
)
### END SOLUTION

print(report_line)
```

```python
tutor.check("ch03_todo_05", report_line)
```

## To-Do 3.6 (Simulated `input()` Hygiene): Clinic Intake & Zero-Padded Tag

A clinic reception form collected four raw strings (simulating `input()`):
`raw_name = "  fatema akter  "`, `raw_age = " 28 "`, `raw_seq = " 7 "`, `raw_temp = " 38.6 C "`.

Compute:
- `clean_name`: `"Fatema Akter"` (`str`)
- `age_int`: `28` (`int`)
- `temp_c`: `38.6` (`float`)
- `case_tag`: `"CASE-0007"` (`str`, zero-padded to 4 digits with `:04d`)

```python
# Simulated raw input() strings
raw_name = "  fatema akter  "
raw_age = " 28 "
raw_seq = " 7 "
raw_temp = " 38.6 C "

# STUB: clean_name = None
# STUB: age_int = None
# STUB: temp_c = None
# STUB: case_tag = None
### BEGIN SOLUTION
clean_name = raw_name.strip().title()
age_int = int(raw_age.strip())
temp_c = float(raw_temp.replace("C", "").strip())
case_tag = f"CASE-{int(raw_seq.strip()):04d}"
### END SOLUTION

print(f"{case_tag} | {clean_name} ({age_int} yrs) | Temp: {temp_c} C")
```

```python
tutor.check("ch03_todo_06", (clean_name, age_int, temp_c, case_tag))
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 3.1 (Silent Immutability Trap):** Writing `raw_district.strip()` without assigning the result throws away the cleaned string—always capture the return value with `clean_district = ...`.
- **To-Do 3.2 (Off-by-One Slicing):** In `"BD-DHK-2026-0482"`, `patient_id[3:5]` returns `"DH"` (2 characters), not `"DHK"`, because Python slices are half-open (`stop` is excluded). Always verify `stop - start == desired_length` (`6 - 3 = 3`).
- **To-Do 3.5 (`:.2%` vs. `* 100`):** Never multiply `prevalence` by `100` when using the `%` format specifier (`{prevalence * 100:.2%}` would print `310.04%` instead of `3.10%`!).
