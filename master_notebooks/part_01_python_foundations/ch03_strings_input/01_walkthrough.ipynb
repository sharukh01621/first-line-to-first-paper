## Section 3.3.1 — Indexing, Negative Indexing & Half-Open Slicing

```python
patient_id = "BD-DHK-2026-0042"
print("Length:", len(patient_id))
print("First char [0]:", patient_id[0], "| Last char [-1]:", patient_id[-1])
print("Country [:2]:", patient_id[:2])
print("District [3:6]:", patient_id[3:6])
print("Year [7:11]:", patient_id[7:11])
print("Serial [-4:]:", patient_id[-4:])
```

## Section 3.3.2 — String Immutability & Method Chaining

```python
raw_clinic = "   dhaka_medical_college   "

# Calling a method WITHOUT reassignment discards the cleaned copy!
raw_clinic.strip()
print("Unchanged raw_clinic:", repr(raw_clinic))

# Capture the new string object with assignment + method chaining
clean_clinic = raw_clinic.strip().replace("_", " ").title()
print("Cleaned clinic:", repr(clean_clinic))
print("Starts with 'Dhaka'?:", clean_clinic.startswith("Dhaka"))
print("Count of 'a':", clean_clinic.lower().count("a"))
```

## Section 3.3.3 — Splitting Delimited Records & Joining Fields

```python
raw_row = "  PAT-019 | 31 | 10.4 g/dL | positive  "
parts = raw_row.split("|")

pat_id = parts[0].strip()
age = int(parts[1].strip())
hb = float(parts[2].replace("g/dL", "").strip())
status = parts[3].strip().upper()

# Reassemble a clean CSV row with .join()
clean_csv = ",".join([pat_id, str(age), str(hb), status])
print("Clean CSV row:", clean_csv)
```

## Section 3.3.5 — Interactive `input()` Hygiene (Simulated for Reproducible Notebooks)

In Python, `input("Prompt: ")` **always returns a `str`**, even when a user types digits. Below, we simulate a clinician's raw text entry so this notebook executes cleanly from top to bottom without blocking.

```python
# In an interactive Colab session, you can uncomment the line below:
# raw_temp = input("Enter patient temperature (e.g., 38.4 C): ")

# For reproducible, non-blocking execution, we simulate the user's text:
raw_temp = "  38.4 C  "
print("Raw input type:", type(raw_temp), "| Value:", repr(raw_temp))

temp_c = float(raw_temp.replace("C", "").strip())
has_fever = temp_c >= 38.0
print(f"Parsed Temp: {temp_c:.1f} C | Fever Flag: {has_fever}")
```

## Section 3.4 — Worked Problem-Structuring Blueprint: Messy Clinical SMS Triage Parser

```python
# Step 1: Raw incoming clinic SMS text
raw_sms = "  barishal_sadar | 007 | 39.1 C | 112000 /uL  "

# Step 2: Split on '|' and sanitize each text field
parts = raw_sms.split("|")
upazila = parts[0].strip().replace("_", " ").title()
serial = int(parts[1].strip())
temp_c = float(parts[2].replace("C", "").strip())
platelets = int(parts[3].replace("/uL", "").strip())

# Step 3: Derive clinical triage flags & standardized ID
case_id = f"BRS-{serial:04d}"
is_febrile = temp_c >= 38.0
low_platelets = platelets < 150000
urgent_referral = is_febrile and low_platelets

# Step 4: Format publication-clean clinical alert
print(f"[{case_id}] {upazila} | Temp: {temp_c:.1f} C")
print(f"Platelets: {platelets:,} /uL | Urgent Referral: {urgent_referral}")
```
