## Exam Task 3.1 (30 Points): Clinical Specimen Barcode Parser

A surveillance laboratory scans a messy barcode string:
`raw_barcode = "  bd-ctg-2026-0094-mal  "`

Standardize and extract with exact required types:
1. `clean_barcode`: stripped and uppercased (`"BD-CTG-2026-0094-MAL"`, `str`)
2. `division_code`: 3-letter division code (`"CTG"`, `str`)
3. `sample_year`: 4-digit collection year (`2026`, `int`)
4. `sample_number`: integer specimen sequence number (`94`, `int`)
5. `is_malaria_panel`: `True` if `clean_barcode` ends with `"MAL"` (`bool`)

```python
raw_barcode = "  bd-ctg-2026-0094-mal  "

# STUB: clean_barcode = None
# STUB: division_code = None
# STUB: sample_year = None
# STUB: sample_number = None
# STUB: is_malaria_panel = None
### BEGIN SOLUTION
clean_barcode = raw_barcode.strip().upper()
tokens = clean_barcode.split("-")
division_code = tokens[1]
sample_year = int(tokens[2])
sample_number = int(tokens[3])
is_malaria_panel = clean_barcode.endswith("MAL")
### END SOLUTION

print(f"{clean_barcode} -> Div={division_code}, Year={sample_year}, #{sample_number}, Malaria={is_malaria_panel}")
```

## Exam Task 3.2 (35 Points): Dirty Field-Survey Intake Cleaner

A mobile data-collection form sends a single semicolon-delimited text payload:
`raw_payload = "  rahim_uddin ; 45 yrs ; 72.4 kg ; 168.0 cm ; 148/92 mmHg  "`

Parse and compute:
1. `patient_name`: stripped, underscores replaced with spaces, Title Case (`"Rahim Uddin"`, `str`)
2. `age_years`: integer age (`45`, `int`)
3. `bmi`: float BMI `weight_kg / ((height_cm / 100) ** 2)` rounded to 2 decimal places (`25.65`, `float`)
4. `sbp_mmhg`: systolic blood pressure (before `"/"`) as `int` (`148`)
5. `dbp_mmhg`: diastolic blood pressure (after `"/"`) as `int` (`92`)

```python
raw_payload = "  rahim_uddin ; 45 yrs ; 72.4 kg ; 168.0 cm ; 148/92 mmHg  "

# STUB: patient_name = None
# STUB: age_years = None
# STUB: bmi = None
# STUB: sbp_mmhg = None
# STUB: dbp_mmhg = None
### BEGIN SOLUTION
fields = raw_payload.split(";")
patient_name = fields[0].strip().replace("_", " ").title()
age_years = int(fields[1].replace("yrs", "").strip())
weight_kg = float(fields[2].replace("kg", "").strip())
height_m = float(fields[3].replace("cm", "").strip()) / 100
bmi = round(weight_kg / (height_m ** 2), 2)
bp_parts = fields[4].replace("mmHg", "").strip().split("/")
sbp_mmhg = int(bp_parts[0])
dbp_mmhg = int(bp_parts[1])
### END SOLUTION

print(f"{patient_name} ({age_years} yrs) | BMI: {bmi} | BP: {sbp_mmhg}/{dbp_mmhg} mmHg")
```

## Exam Task 3.3 (35 Points): Publication-Ready Surveillance Bulletin Generator

Given district outbreak summary variables:
`facility_code = "uhc-syl-03"`, `report_seq = 14`, `population = 340500`, `confirmed_cases = 512`, `deaths = 9`.

Compute and format:
1. `report_id`: uppercase facility code + `"-R"` + zero-padded 4-digit `report_seq` (`"UHC-SYL-03-R0014"`)
2. `incidence_per_100k`: `round((confirmed_cases / population) * 100000, 2)` (`150.37`)
3. `cfr_pct_str`: Case Fatality Rate (`deaths / confirmed_cases`) formatted as a 2-decimal percentage string (`"1.76%"`)
4. `bulletin_line`: exact formatted summary string:  
   `"[UHC-SYL-03-R0014] Pop=340,500 | Cases=512 (150.4/100k) | CFR=1.76%"`

```python
facility_code = "uhc-syl-03"
report_seq = 14
population = 340500
confirmed_cases = 512
deaths = 9

# STUB: report_id = None
# STUB: incidence_per_100k = None
# STUB: cfr_pct_str = None
# STUB: bulletin_line = None
### BEGIN SOLUTION
report_id = f"{facility_code.upper()}-R{report_seq:04d}"
incidence_per_100k = round((confirmed_cases / population) * 100000, 2)
cfr_pct_str = f"{deaths / confirmed_cases:.2%}"
bulletin_line = (
    f"[{report_id}] Pop={population:,} | "
    f"Cases={confirmed_cases:,} ({incidence_per_100k:.1f}/100k) | "
    f"CFR={cfr_pct_str}"
)
### END SOLUTION

print(bulletin_line)
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 3.1 (Splitting vs. Hard-Coded Slicing):** Calling `.strip().upper()` *before* `.split("-")` guarantees uniform uppercase tokens and prevents hidden trailing spaces from breaking `.endswith("MAL")`.
- **Task 3.2 (Stripping Units Before Casting):** Calling `float("72.4 kg")` raises `ValueError` because of the unit suffix `"kg"`. Removing units with `.replace(...)` followed by `.strip()` ensures clean numeric conversion—and splitting `"148/92"` on `"/"` cleanly separates systolic and diastolic readings.
- **Task 3.3 (Zero-Padding & Percentage Formatters):** `{report_seq:04d}` pads `14` to `0014` without manual string concatenation, and `{deaths / confirmed_cases:.2%}` converts the decimal proportion `0.017578...` directly into `"1.76%"`.
