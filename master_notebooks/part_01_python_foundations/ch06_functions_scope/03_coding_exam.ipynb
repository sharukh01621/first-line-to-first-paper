## Exam Task 6.1 (30 Points): District Name Normalizer

Write a pure function `normalize_district_name(raw_name)` that cleans messy district names from field surveys:
1. If `raw_name` is `None`, return `"Unknown"`.
2. Strip leading/trailing whitespace and collapse any extra internal spaces/tabs/newlines into a single space.
3. If the cleaned string is empty (`""`), return `"Unknown"`.
4. Otherwise return the cleaned name in **Title Case** (e.g., `"  dhaka   north  "` $\rightarrow$ `"Dhaka North"`).

```python
def normalize_district_name(raw_name):
    """Normalize messy district strings to Title Case, or 'Unknown'."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    if raw_name is None:
        return "Unknown"
    cleaned = " ".join(raw_name.split())
    if not cleaned:
        return "Unknown"
    return cleaned.title()
    ### END SOLUTION
```

## Exam Task 6.2 (35 Points): Safe $2 \times 2$ Odds Ratio Function

Write `compute_odds_ratio(a, b, c, d)` for a standard epidemiology $2 \times 2$ table:
- If `b == 0` or `c == 0`, return `None` (preventing `ZeroDivisionError`).
- Otherwise compute the exact cross-product ratio `(a * d) / (b * c)` and **return** it rounded to **2 decimal places**.

```python
def compute_odds_ratio(a, b, c, d):
    """Return 2x2 Odds Ratio (2 decimals), or None if b or c is 0."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    if b == 0 or c == 0:
        return None
    return round((a * d) / (b * c), 2)
    ### END SOLUTION
```

## Exam Task 6.3 (35 Points): Clinic Batch Hemoglobin Summarizer

Write `summarize_clinic_batch(hb_readings)` that processes a list of maternal hemoglobin readings (g/dL):
1. Skip all `None` values and all sentinel `999` values.
2. If no valid readings remain, return `(0, 0, None)`.
3. Otherwise return a **3-tuple** `(valid_count, anemic_count, mean_hb)` where:
   - `valid_count` (`int`): count of valid readings
   - `anemic_count` (`int`): count of valid readings strictly `< 11.0`
   - `mean_hb` (`float`): arithmetic mean of valid readings rounded to **2 decimal places**

```python
def summarize_clinic_batch(hb_readings):
    """Return (valid_count, anemic_count, mean_hb) skipping None and 999."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    valid_count = 0
    anemic_count = 0
    total_hb = 0.0
    for val in hb_readings:
        if val is None or val == 999:
            continue
        valid_count += 1
        total_hb += val
        if val < 11.0:
            anemic_count += 1
    if valid_count == 0:
        return (0, 0, None)
    mean_hb = round(total_hb / valid_count, 2)
    return (valid_count, anemic_count, mean_hb)
    ### END SOLUTION
```

- **Task 6.2 (Zero Cells & Parentheses):** Remember that `a == 0` produces a valid Odds Ratio of `0.0`—only `b == 0` or `c == 0` causes division by zero. Always wrap `(b * c)` in parentheses so Python divides by the full denominator.
- **Task 6.3 (Boundary `< 11.0` & Empty-Batch Return):** A reading of exactly `11.0` g/dL is **Normal** (not `< 11.0`). Returning `(0, 0, None)` when `valid_count == 0` keeps the 3-tuple return contract consistent for caller unpacking.
