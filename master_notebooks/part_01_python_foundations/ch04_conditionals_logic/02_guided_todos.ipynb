## To-Do 4.3 (Boolean Algebra): Vaccine Trial Eligibility Screen

A trial volunteer has `age_years = 24`, `is_pregnant = False`, `has_severe_allergy = False`, and `consented = True`. Compute:
- `age_ok`: `18 <= age_years <= 65`
- `medically_safe`: `True` only if NOT pregnant AND does NOT have a severe allergy
- `can_enroll`: `age_ok and medically_safe and consented`
- `needs_review`: `is_pregnant or has_severe_allergy`

```python
age_years = 24
is_pregnant = False
has_severe_allergy = False
consented = True

# STUB: age_ok = None
# STUB: medically_safe = None
# STUB: can_enroll = None
# STUB: needs_review = None
### BEGIN SOLUTION
age_ok = 18 <= age_years <= 65
medically_safe = (not is_pregnant) and (not has_severe_allergy)
can_enroll = age_ok and medically_safe and consented
needs_review = is_pregnant or has_severe_allergy
### END SOLUTION

print(f"Enroll: {can_enroll} | Review: {needs_review}")
```

## To-Do 4.5 (Short-Circuit Guard): Avoiding `ZeroDivisionError`

A surveillance ward reports `cases = 15` and `population = 0` (census count missing). Safely compute:
- `has_valid_pop`: `population > 0`
- `is_high_incidence`: `population > 0 and (cases / population) * 100000 >= 50.0`
- `incidence_per_100k`: `round((cases / population) * 100000, 2)` if `population > 0` else `None`

```python
cases = 15
population = 0

# STUB: has_valid_pop = None
# STUB: is_high_incidence = None
# STUB: incidence_per_100k = "NOT_SET"
### BEGIN SOLUTION
has_valid_pop = population > 0
is_high_incidence = (
    population > 0 and (cases / population) * 100000 >= 50.0
)
if population > 0:
    incidence_per_100k = round((cases / population) * 100000, 2)
else:
    incidence_per_100k = None
### END SOLUTION

print(has_valid_pop, is_high_incidence, incidence_per_100k)
```

```python
tutor.check("ch04_todo_05", (has_valid_pop, is_high_incidence, incidence_per_100k))
```

## To-Do 4.6 (Stretch Guard & Triage): Intercepting Missing Sentinel `999` / `None`

A clinic intake record has `raw_hb = 999` (DHS missing-value sentinel!) and `sbp_mmhg = 148`. Compute:
- `hb_missing`: `True` if `raw_hb is None or raw_hb == 999`
- `anemia_status`: `"Missing"` if `hb_missing`, else `"Anemic"` if `raw_hb < 11.0`, else `"Normal"`
- `triage_priority`:
  - `"Urgent"` if `sbp_mmhg >= 160` or (`not hb_missing` and `raw_hb < 7.0`)
  - `"Moderate"` if `sbp_mmhg >= 140` or `hb_missing`
  - `"Routine"` otherwise

```python
raw_hb = 999
sbp_mmhg = 148

# STUB: hb_missing = None
# STUB: anemia_status = None
# STUB: triage_priority = None
### BEGIN SOLUTION
hb_missing = (raw_hb is None) or (raw_hb == 999)

if hb_missing:
    anemia_status = "Missing"
elif raw_hb < 11.0:
    anemia_status = "Anemic"
else:
    anemia_status = "Normal"

if sbp_mmhg >= 160 or (not hb_missing and raw_hb < 7.0):
    triage_priority = "Urgent"
elif sbp_mmhg >= 140 or hb_missing:
    triage_priority = "Moderate"
else:
    triage_priority = "Routine"
### END SOLUTION

print(f"Missing: {hb_missing} | Status: {anemia_status} | Triage: {triage_priority}")
```

```python
tutor.check("ch04_todo_06", (hb_missing, anemia_status, triage_priority))
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 4.4 (`if` vs `elif` overwrite bug):** Writing three separate `if` statements causes `hb_g_dl = 8.9` to trigger `hb_g_dl < 10.0` (`"Moderate"`) and then immediately get overwritten by `hb_g_dl < 11.0` (`"Mild"`). Always use `elif` for mutually exclusive clinical categories.
- **To-Do 4.5 (Short-circuit guard order):** Python evaluates `A and B` left-to-right and skips `B` whenever `A` is `False`. Putting `population > 0` on the left shields `cases / population` from `ZeroDivisionError`.
- **To-Do 4.6 (Survey sentinel `999`):** Without checking `raw_hb == 999` *before* `< 11.0`, a missing blood test silently looks like a super-healthy `999 g/dL` reading (`"Normal"`)!
