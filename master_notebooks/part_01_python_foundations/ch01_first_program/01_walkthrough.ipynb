## Section 1.3.2 — Python as a Scientific & Epidemiological Calculator

```python
# Basic arithmetic operators in Python
print("Addition (+):", 120 + 45)
print("Subtraction (-):", 120 - 45)
print("Multiplication (*):", 14 * 3)
print("True Division (/):", 10 / 4)        # Always returns a decimal (float): 2.5
print("Floor Division (//):", 10 // 4)    # Whole groups only: 2
print("Remainder / Modulo (%):", 10 % 4)  # Leftover after packing: 2
print("Exponentiation (**):", 1.65 ** 2)  # Height squared for BMI: 2.7225
```

```python
# Operator precedence: Why parentheses save lives in biostatistics!
wrong_mean = 120 + 130 + 140 / 3
right_mean = (120 + 130 + 140) / 3

print("Without parentheses (BUG!):", wrong_mean)
print("With parentheses (CORRECT):", right_mean)
```
