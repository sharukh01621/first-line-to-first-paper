## Section 2.3.1 — Variables as Name Tags (`id()` and Rebinding)

```python
# A variable is a name tag attached to an object in memory
cases = 1000
print("Initial value:", cases, "| Memory ID:", id(cases))

# Rebinding attaches the tag 'cases' to a BRAND NEW integer object (1250)
cases = cases + 250
print("Updated value:", cases, "| Memory ID:", id(cases))
```

## Section 2.3.2 — The 4 Core Primitive Types + `None` (Missing Sentinel)

```python
patient_id = "UHC-0042"   # str  (text)
age_years = 29             # int  (exact whole number)
hb_g_dl = 10.6             # float (decimal measurement)
is_anemic = hb_g_dl < 11.0 # bool (True or False)
follow_up_date = None      # NoneType (explicitly missing / not yet recorded)

print(patient_id,     "->", type(patient_id))
print(age_years,      "->", type(age_years))
print(hb_g_dl,        "->", type(hb_g_dl))
print(is_anemic,      "->", type(is_anemic))
print(follow_up_date, "->", type(follow_up_date))
```

## Section 2.3.3 — Augmented Assignment (`+=`, `-=`, `*=`, `/=`)

```python
vaccine_stock = 500
vaccine_stock -= 85   # 85 doses administered in morning clinic -> 415
vaccine_stock += 200  # District cold-chain resupply arrives    -> 615
print("End-of-day vaccine doses:", vaccine_stock)
```

## Section 2.3.4 — Explicit Type Casting for Messy Survey Strings

```python
# Raw CSV rows always arrive as strings!
raw_age = "34"
raw_temp = "38.5"
raw_months = "24.0"

age_int = int(raw_age)
temp_float = float(raw_temp)
months_int = int(float(raw_months))  # Two-step idiom for decimal-strings!

print(age_int, type(age_int))
print(temp_float, type(temp_float))
print(months_int, type(months_int))
```

## Section 2.3.5 — Solving the Chapter 1 Mystery: Binary Floating-Point Precision

```python
import math

print("0.1 + 0.2      =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3 ->", (0.1 + 0.2) == 0.3)
print("math.isclose()   ->", math.isclose(0.1 + 0.2, 0.3))

# Paying off Chapter 1's ORS supply mystery:
print("420 * 1.15       =", 420 * 1.15)
print("int(420 * 1.15)  =", int(420 * 1.15), "  <-- chops decimal!")
print("round(420 * 1.15)=", round(420 * 1.15), " <-- exact nearest integer!")
```

## Section 2.4 — Worked Problem-Structuring Blueprint: Raw Intake Parser & Priority Triage

```python
# Step 1: Raw CSV-like string inputs
raw_weight_kg = "64.5"
raw_height_cm = "168"
raw_hb = "10.4"
raw_temp_c = "38.2"

# Step 2: Explicit type normalization & unit conversion
weight_kg = float(raw_weight_kg)
height_m = float(raw_height_cm) / 100
hb_g_dl = float(raw_hb)
temp_c = float(raw_temp_c)

# Step 3: Derived metrics & boolean clinical screening flags
bmi = round(weight_kg / (height_m ** 2), 2)
is_febrile = temp_c >= 38.0
is_anemic = hb_g_dl < 11.0
needs_priority_review = is_febrile or is_anemic

# Step 4: Clean clinical triage summary
print(f"BMI: {bmi} kg/m^2 | Febrile: {is_febrile} | Anemic: {is_anemic}")
print(f"Priority Review Needed: {needs_priority_review}")
```
