## To-Do 2.5 (Scientific Precision): `==` vs. `math.isclose()`

A lab protocol mixes `dose_a = 0.1` mL and `dose_b = 0.2` mL to hit `target_dose = 0.3` mL.  
Compute:
- `exact_eq`: `(dose_a + dose_b) == target_dose`
- `safe_eq`: `math.isclose(dose_a + dose_b, target_dose)`

```python
import math

dose_a = 0.1
dose_b = 0.2
target_dose = 0.3

# STUB: exact_eq = None
# STUB: safe_eq = None
### BEGIN SOLUTION
exact_eq = (dose_a + dose_b) == target_dose
safe_eq = math.isclose(dose_a + dose_b, target_dose)
### END SOLUTION

print(f"Exact == : {exact_eq} | Safe math.isclose : {safe_eq}")
```

```python
tutor.check("ch02_todo_05", (exact_eq, safe_eq))
```

## To-Do 2.6 (Stretch Clinical Math): Pediatric Weight-Based Dosage

A child weighs `weight_kg = 18.5` kg. Standard dosing is `15` mg/kg per dose, given `3` times daily. Each tablet contains `250` mg.  
Compute:
- `daily_mg`: total milligrams per day
- `exact_tablets`: `daily_mg / 250`, rounded to **2 decimal places**
- `dose_is_float`: `isinstance(daily_mg, float)`

```python
weight_kg = 18.5
mg_per_kg = 15
doses_per_day = 3
tablet_mg = 250

# STUB: daily_mg = None
# STUB: exact_tablets = None
# STUB: dose_is_float = None
### BEGIN SOLUTION
daily_mg = weight_kg * mg_per_kg * doses_per_day
exact_tablets = round(daily_mg / tablet_mg, 2)
dose_is_float = isinstance(daily_mg, float)
### END SOLUTION

print(f"Daily dose: {daily_mg} mg ({exact_tablets} tablets) | Float? {dose_is_float}")
```

```python
tutor.check("ch02_todo_06", (daily_mg, exact_tablets, dose_is_float))
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 2.4 (`int("24.0")` crash):** Python's `int()` only parses strings made purely of integer digits (plus optional sign). Because SPSS/Stata/Pandas CSV exports frequently serialize integer columns with missing values as floats (`"24.0"`), `int(float(raw_months))` is essential.
- **To-Do 2.5 (`0.1 + 0.2 == 0.3` is `False`):** Never test floating-point equality with `==` in statistical or ML assertions—always use `math.isclose()`.
