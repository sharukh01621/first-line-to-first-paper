- `.mean_bmi(self)`: return the mean of `p.bmi()` across all patients rounded to **2 decimal places**, or `None` if the cohort is empty.

```python
class ClinicCohort:
    """Container managing a cohort of patient records."""

    def __init__(self, name):
        # STUB: raise NotImplementedError("Replace with your solution")
        ### BEGIN SOLUTION
        self.name = name
        self.patients = []

    def add_patient(self, patient):
        self.patients.append(patient)

    def __len__(self):
        return len(self.patients)

    def mean_bmi(self):
        if not self.patients:
            return None
        total_bmi = sum(p.bmi() for p in self.patients)
        return round(total_bmi / len(self.patients), 2)
        ### END SOLUTION
```

## To-Do 12.4 (Scikit-Learn Estimator Pattern): `MeanImputer`

Implement a Scikit-Learn style `MeanImputer` transformer:
- `__init__(self)`: initialize `self.mean_ = None`.
- `.fit(self, values)`: compute the mean of valid readings (skipping `None` and `999`), store `self.mean_` rounded to **2 decimal places**, and `return self` (without mutating `values`).
- `.transform(self, values)`: return a **new list** replacing every `None` or `999` with `self.mean_`, leaving valid values intact and **never mutating `values`**.

```python
class MeanImputer:
    """Scikit-Learn style mean imputer skipping None and 999."""

    def __init__(self):
        # STUB: raise NotImplementedError("Replace with your solution")
        ### BEGIN SOLUTION
        self.mean_ = None

    def fit(self, values):
        valid = [v for v in values if v is not None and v != 999]
        self.mean_ = round(sum(valid) / len(valid), 2)
        return self

    def transform(self, values):
        return [
            self.mean_ if (v is None or v == 999) else v
            for v in values
        ]
        ### END SOLUTION
```

## To-Do 12.5 (Scikit-Learn Feature Scaling): `MinMaxScaler`

Implement a Scikit-Learn style `MinMaxScaler` transformer:
- `__init__(self)`: initialize `self.min_ = None` and `self.max_ = None`.
- `.fit(self, values)`: learn `self.min_ = min(values)` and `self.max_ = max(values)`, then `return self`.
- `.transform(self, values)`: return a **new list** scaling each `x` to `[0.0, 1.0]` via `round((x - self.min_) / (self.max_ - self.min_), 4)`. If `self.max_ == self.min_`, return `0.0` for every entry without raising `ZeroDivisionError`. Never mutate `values`.

```python
class MinMaxScaler:
    """Scikit-Learn style [0, 1] feature scaler with constant guard."""

    def __init__(self):
        # STUB: raise NotImplementedError("Replace with your solution")
        ### BEGIN SOLUTION
        self.min_ = None
        self.max_ = None

    def fit(self, values):
        self.min_ = min(values)
        self.max_ = max(values)
        return self

    def transform(self, values):
        span = self.max_ - self.min_
        if span == 0:
            return [0.0 for _ in values]
        return [round((x - self.min_) / span, 4) for x in values]
        ### END SOLUTION
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 12.2 (Mutable Class Attribute Trap):** Writing `patients = []` in the class body instead of `self.patients = []` inside `__init__` causes EVERY `ClinicCohort` instance to share one global list!
- **To-Do 12.4 & 12.5 (Train/Test Anti-Leakage):** Never recompute `mean`, `min`, or `max` inside `.transform(values)`! Fit learns parameters strictly from training data (`self.mean_`, `self.min_`, `self.max_`), and `.transform()` reuses those frozen parameters on unseen test splits without mutating inputs.
- **To-Do 12.6 (`__call__` Delegation):** In PyTorch, you always write `model(x)` instead of `model.forward(x)` because `nn.Module.__call__` executes hooks and then delegates to `self.forward(x)`.
