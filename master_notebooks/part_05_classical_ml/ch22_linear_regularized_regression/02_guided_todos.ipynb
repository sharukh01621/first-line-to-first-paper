## To-Do 22.1 (Regression Metrics): `regression_metrics(y_true, y_pred)`

Implement `regression_metrics(y_true, y_pred)` without mutating caller inputs:
- Compute Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and Coefficient of Determination ($R^2 = 1 - \text{SS}_{\text{res}} / \text{SS}_{\text{tot}}$).
- Return `{"mae": round4, "rmse": round4, "r2": round4}` with each metric rounded to **4 decimal places**.

```python
def regression_metrics(y_true, y_pred):
    """Return dict with MAE, RMSE, and R^2 rounded to 4 decimals."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    yt = np.asarray(y_true, dtype=float)
    yp = np.asarray(y_pred, dtype=float)
    err = yt - yp
    mae = float(np.mean(np.abs(err)))
    rmse = float(np.sqrt(np.mean(err ** 2)))
    ss_res = float(np.sum(err ** 2))
    ss_tot = float(np.sum((yt - np.mean(yt)) ** 2))
    r2 = 1.0 - ss_res / ss_tot
    return {
        "mae": round(mae, 4),
        "rmse": round(rmse, 4),
        "r2": round(r2, 4),
    }
    ### END SOLUTION
```

## To-Do 22.2 (Closed-Form Normal Equation): `ols_normal_equation(X, y)`

Implement pure-NumPy OLS `ols_normal_equation(X, y)`:
- Prepend an intercept column of `1`s to the 2D matrix `X` (`Xb = np.column_stack([np.ones(len(X)), X])`).
- Solve safely via `np.linalg.lstsq(Xb, y, rcond=None)[0]` (or `np.linalg.pinv(Xb.T @ Xb) @ Xb.T @ y`) so collinear matrices never crash with `LinAlgError`.
- Return `(float_intercept, 1D_coef_array)` without mutating `X` or `y`.

```python
def ols_normal_equation(X, y):
    """Solve OLS via Normal Equation / lstsq; return (intercept, coef)."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    X_arr = np.asarray(X, dtype=float)
    y_arr = np.asarray(y, dtype=float)
    Xb = np.column_stack([np.ones(len(X_arr)), X_arr])
    theta = np.linalg.lstsq(Xb, y_arr, rcond=None)[0]
    return (float(theta[0]), theta[1:])
    ### END SOLUTION
```

## To-Do 22.3 (Batch Gradient Descent): `gradient_descent_linear(X, y, lr=0.05, epochs=300)`

Implement pure-NumPy batch gradient descent minimizing Mean Squared Error:
- Initialize `w = np.zeros(p)` and `b = 0.0`.
- For each epoch, compute `err = (X @ w + b) - y`, append `float(np.mean(err ** 2))` to `loss_history`, and update:
  $$w \leftarrow w - \eta \cdot \frac{2}{n} X^T (\hat{y} - y), \qquad b \leftarrow b - \eta \cdot \frac{2}{n} \sum_{i=1}^n (\hat{y}_i - y_i)$$
- Return `(w_1d, b_float, loss_history_list)` with `len(loss_history) == epochs` and `loss_history[-1] < loss_history[0]`.

```python
def gradient_descent_linear(X, y, lr=0.05, epochs=300):
    """Run batch gradient descent on MSE; return (w, b, loss_history)."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    X_arr = np.asarray(X, dtype=float)
    y_arr = np.asarray(y, dtype=float)
    n, p = X_arr.shape
    w = np.zeros(p, dtype=float)
    b = 0.0
    loss_history = []
    for _ in range(epochs):
        err = (X_arr @ w + b) - y_arr
        loss_history.append(float(np.mean(err ** 2)))
        dw = (2.0 / n) * (X_arr.T @ err)
        db = (2.0 / n) * float(np.sum(err))
        w = w - lr * dw
        b = b - lr * db
    return (w, float(b), loss_history)
    ### END SOLUTION
```

## To-Do 22.4 (Leak-Free Ridge Pipeline): `fit_ridge_standardized(X_train, y_train, X_test, alpha=1.0)`

Implement `fit_ridge_standardized(X_train, y_train, X_test, alpha=1.0)`:
- Build `Pipeline([("scaler", StandardScaler()), ("ridge", Ridge(alpha=alpha))])`.
- Fit on `(X_train, y_train)` only, predict on `X_test`, and extract `pipe.named_steps["ridge"].coef_`.
- Return `(y_pred_test, coef_array)`.

```python
def fit_ridge_standardized(X_train, y_train, X_test, alpha=1.0):
    """Fit leak-free StandardScaler + Ridge; return (y_pred, coef)."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=alpha)),
    ])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    return (y_pred, pipe.named_steps["ridge"].coef_)
    ### END SOLUTION
```

## To-Do 22.5 (Lasso Feature Selection): `count_lasso_selected_features(X_train, y_train, alpha=0.1, tol=1e-6)`

Implement `count_lasso_selected_features(X_train, y_train, alpha=0.1, tol=1e-6)`:
- Fit a standardized `Pipeline([("scaler", StandardScaler()), ("lasso", Lasso(alpha=alpha, max_iter=5000))])` on `(X_train, y_train)`.
- Return the Python `int` count of coefficients satisfying `abs(coef) > tol`.

```python
def count_lasso_selected_features(
    X_train, y_train, alpha=0.1, tol=1e-6
):
    """Count standardized Lasso coefficients with |coef| > tol."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("lasso", Lasso(alpha=alpha, max_iter=5000)),
    ])
    pipe.fit(X_train, y_train)
    coefs = pipe.named_steps["lasso"].coef_
    return int(np.sum(np.abs(coefs) > tol))
    ### END SOLUTION
```

```python
tutor.check("ch22_todo_05", count_lasso_selected_features)
```

## To-Do 22.6 (Residual Diagnostics): `residual_summary(y_true, y_pred)`

Implement `residual_summary(y_true, y_pred)`:
- Define residuals as `resid = y_true - y_pred` (actual minus predicted).
- Compute mean residual, standard deviation (`np.std(resid)`, default `ddof=0`), and maximum absolute error (`np.max(np.abs(resid))`).
- Return `{"mean_residual": round4, "std_residual": round4, "max_abs_error": round4}`.

```python
def residual_summary(y_true, y_pred):
    """Return mean, std (ddof=0), and max |error| of y_true - y_pred."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    resid = (
        np.asarray(y_true, dtype=float)
        - np.asarray(y_pred, dtype=float)
    )
    return {
        "mean_residual": round(float(np.mean(resid)), 4),
        "std_residual": round(float(np.std(resid)), 4),
        "max_abs_error": round(float(np.max(np.abs(resid))), 4),
    }
    ### END SOLUTION
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 22.2 (`np.linalg.inv` vs. `np.linalg.lstsq`):** Naive inversion of `Xb.T @ Xb` crashes with `LinAlgError: Singular matrix` whenever two predictors are perfectly collinear. Using `np.linalg.lstsq(Xb, y, rcond=None)` (or `np.linalg.pinv`) computes the minimum-norm least-squares solution cleanly.
- **To-Do 22.3 (Gradient Sign & Unscaled Learning Rate):** Flipping `err = y_hat - y` to `y - y_hat` without flipping the update sign causes gradient *ascent*, exploding the MSE to `inf`/`nan`. Always verify `loss_history[-1] < loss_history[0]`.
- **To-Do 22.4 & 22.5 (Scaling Inside `Pipeline`):** Penalizing raw unstandardized features makes `Ridge` and `Lasso` arbitrarily depend on measurement units (e.g., meters vs. millimeters). Wrapping `StandardScaler()` inside a `Pipeline` enforces equal-footing regularization and eliminates train/test leakage.
