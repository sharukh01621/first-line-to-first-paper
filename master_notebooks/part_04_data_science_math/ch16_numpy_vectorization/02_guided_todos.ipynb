## To-Do 16.1 (Vectorized Arithmetic): `vectorized_bmi(weights_kg, heights_cm)`

Implement `vectorized_bmi(weights_kg, heights_cm)`:
- Accept array-like inputs (Python lists or NumPy arrays) without mutating them.
- Convert `heights_cm` to meters (`/ 100.0`) and compute $\text{BMI} = \text{weight} / \text{height}_m^2$ elementwise.
- Return a 1D `np.ndarray` of floats rounded to **2 decimal places** via `np.round(..., 2)`.

```python
def vectorized_bmi(weights_kg, heights_cm):
    """Compute 1D NumPy array of BMIs rounded to 2 decimals."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    w = np.asarray(weights_kg, dtype=float)
    h_m = np.asarray(heights_cm, dtype=float) / 100.0
    return np.round(w / (h_m ** 2), 2)
    ### END SOLUTION
```

## To-Do 16.2 (Masking & `np.nanmean`): `clean_and_mean_hb(hb_array)`

Implement `clean_and_mean_hb(hb_array)`:
- Work on a **float copy** (`np.array(hb_array, dtype=float, copy=True)`) so the caller's array is never mutated.
- Replace sentinel `999` with `np.nan`.
- Return `None` if all values are `NaN` (or the array is empty); otherwise return a native Python `float` rounded to **2 decimal places** using `np.nanmean`.

```python
def clean_and_mean_hb(hb_array):
    """Replace 999 with NaN on a float copy and return nanmean."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    arr = np.array(hb_array, dtype=float, copy=True)
    arr[arr == 999] = np.nan
    if arr.size == 0 or np.all(np.isnan(arr)):
        return None
    return round(float(np.nanmean(arr)), 2)
    ### END SOLUTION
```

## To-Do 16.3 (Broadcasting & `axis=0`): `standardize_columns(X)`

Implement `standardize_columns(X)` for a 2D `(n, p)` feature matrix:
- Work on a float copy without mutating `X`.
- Compute column-wise Z-scores $(X - \mu) / \sigma$ along `axis=0` using broadcasting.
- Guard zero-standard-deviation (constant) columns by setting their output column to `0.0`.

```python
def standardize_columns(X):
    """Column-wise Z-score standardization with zero-std guard."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    arr = np.array(X, dtype=float, copy=True)
    means = np.mean(arr, axis=0)
    stds = np.std(arr, axis=0)
    safe_stds = np.where(stds == 0, 1.0, stds)
    z = (arr - means) / safe_stds
    z[:, stds == 0] = 0.0
    return z
    ### END SOLUTION
```

## To-Do 16.4 (Boolean Masking): `flag_fever_mask(temps_c, cutoff=38.0)`

Implement `flag_fever_mask(temps_c, cutoff=38.0)`:
- Convert `temps_c` to a float NumPy array.
- Build a boolean mask where `temps >= cutoff` (inclusive).
- Return a 2-tuple `(mask, count)` where `mask` is a boolean `np.ndarray` and `count` is a native Python `int`.

```python
def flag_fever_mask(temps_c, cutoff=38.0):
    """Return (bool_mask, int_count) for temperatures >= cutoff."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    arr = np.asarray(temps_c, dtype=float)
    mask = arr >= cutoff
    return mask, int(np.sum(mask))
    ### END SOLUTION
```

## To-Do 16.5 (Matrix Multiply `@`): `weighted_risk_scores(X, weights, bias)`

Implement `weighted_risk_scores(X, weights, bias)`:
- Convert `X` `(n, p)` and `weights` `(p,)` with `np.asarray(..., dtype=float)`.
- Compute `X @ weights + bias`.
- Return a 1D `np.ndarray` rounded to **4 decimal places**.

```python
def weighted_risk_scores(X, weights, bias):
    """Compute vectorized linear risk scores X @ weights + bias."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    X_arr = np.asarray(X, dtype=float)
    w_arr = np.asarray(weights, dtype=float)
    return np.round(X_arr @ w_arr + bias, 4)
    ### END SOLUTION
```

## To-Do 16.6 (Broadcasting & Norms): `pairwise_l2_distances(points, reference)`

Implement `pairwise_l2_distances(points, reference)`:
- Given 2D `points` `(n, d)` and 1D `reference` `(d,)`, compute row-wise Euclidean distances using broadcasting subtraction and `np.linalg.norm(points - reference, axis=1)`.
- Return a 1D `np.ndarray` rounded to **4 decimal places**.

```python
def pairwise_l2_distances(points, reference):
    """Return 1D Euclidean distances from each row to reference."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    pts = np.asarray(points, dtype=float)
    ref = np.asarray(reference, dtype=float)
    return np.round(np.linalg.norm(pts - ref, axis=1), 4)
    ### END SOLUTION
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 16.2 (Integer Copy Trap):** Assigning `np.nan` into an integer NumPy array raises `ValueError`, and modifying `hb_array` directly mutates the caller's data. Always copy as float first (`np.array(hb_array, dtype=float, copy=True)`).
- **To-Do 16.3 (`axis=0` vs. `axis=1` & Constant Columns):** In a `(n, p)` feature matrix, `axis=0` collapses down rows to yield one statistic per feature column `(p,)`. Substituting `1.0` where `stds == 0` via `np.where` prevents divide-by-zero warnings and `NaN` propagation.
- **To-Do 16.5 & 16.6 (`@` vs. `*` and Norm Axis):** Using `*` instead of `@` produces an elementwise `(n, p)` array rather than `(n,)` dot products, and omitting `axis=1` in `np.linalg.norm(pts - ref, axis=1)` collapses the entire matrix into a single scalar Frobenius norm.
