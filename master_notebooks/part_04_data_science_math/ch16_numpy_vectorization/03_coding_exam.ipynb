## Exam Task 16.1 (30 Points): `normalize_vital_matrix(X)`

Implement `normalize_vital_matrix(X)` for a 2D numeric matrix `X`:
1. Create a float copy (`np.array(X, dtype=float, copy=True)`) so `X` is never mutated.
2. Replace sentinel `999.0` with `np.nan`.
3. Impute all `np.nan` entries using each column's `np.nanmean(..., axis=0)`.
4. Min-max scale each column to `[0.0, 1.0]`, setting any constant column (`col_max == col_min`) to `0.0`.
5. Return the resulting 2D `np.ndarray` rounded to **4 decimal places**.

```python
def normalize_vital_matrix(X):
    """Clean 999.0, impute column nanmeans, and min-max scale [0, 1]."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    arr = np.array(X, dtype=float, copy=True)
    arr[arr == 999.0] = np.nan
    col_means = np.nanmean(arr, axis=0)
    arr = np.where(np.isnan(arr), col_means, arr)
    col_min = np.min(arr, axis=0)
    span = np.max(arr, axis=0) - col_min
    safe_span = np.where(span == 0, 1.0, span)
    scaled = (arr - col_min) / safe_span
    scaled[:, span == 0] = 0.0
    return np.round(scaled, 4)
    ### END SOLUTION
```

## Exam Task 16.2 (35 Points): `district_weekly_summary(cases_2d, pop_1d)`

Given a 2D surveillance matrix `cases_2d` of shape `(n_districts, n_weeks)` and a 1D population vector `pop_1d` of shape `(n_districts,)`, return a `dict` with these exact 4 keys:
- `"total_per_district"`: 1D `np.ndarray` of total cases per district (`axis=1`).
- `"weekly_mean"`: 1D `np.ndarray` of mean cases per week across districts (`axis=0`), rounded to **2 decimal places**.
- `"peak_week_index"`: 1D integer `np.ndarray` giving the 0-based index of the peak week for each district (`np.argmax(..., axis=1)`).
- `"incidence_per_100k"`: 1D `np.ndarray` of `(total_per_district / pop_1d) * 100_000`, rounded to **2 decimal places**.

```python
def district_weekly_summary(cases_2d, pop_1d):
    """Aggregate district surveillance grid across axes & population."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    cases = np.asarray(cases_2d, dtype=float)
    pop = np.asarray(pop_1d, dtype=float)
    total = np.sum(cases, axis=1)
    weekly_mean = np.round(np.mean(cases, axis=0), 2)
    peak_week = np.argmax(cases, axis=1)
    incidence = np.round((total / pop) * 100_000, 2)
    return {
        "total_per_district": total,
        "weekly_mean": weekly_mean,
        "peak_week_index": peak_week,
        "incidence_per_100k": incidence,
    }
    ### END SOLUTION
```

## Exam Task 16.3 (35 Points): `linear_clinical_scorer(X, weights, bias, threshold=0.5)`

Implement a vectorized logistic clinical risk scorer:
1. Compute linear logits `z = X @ weights + bias` (converting `X` and `weights` via `np.asarray(..., dtype=float)`).
2. Compute sigmoid probabilities `probs = 1.0 / (1.0 + np.exp(-z))`.
3. Compute binary integer class predictions `preds = (probs >= threshold).astype(int)`.
4. Return the 2-tuple `(np.round(probs, 4), preds)`.

```python
def linear_clinical_scorer(X, weights, bias, threshold=0.5):
    """Compute logistic risk probabilities and 0/1 class predictions."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    X_arr = np.asarray(X, dtype=float)
    w_arr = np.asarray(weights, dtype=float)
    z = X_arr @ w_arr + bias
    probs = 1.0 / (1.0 + np.exp(-z))
    preds = (probs >= threshold).astype(int)
    return np.round(probs, 4), preds
    ### END SOLUTION
```

## Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 16.1 (Non-Mutating Copy & Constant Columns):** Converting `999.0` to `np.nan` directly on `X` destroys the caller's matrix, and dividing by `col_max - col_min` on a constant column generates `NaN`s. Use `np.array(X, dtype=float, copy=True)`, impute via `np.where(np.isnan(arr), col_means, arr)`, and zero out `span == 0` columns cleanly.
- **Task 16.2 (`axis=1` vs. `axis=0` on Rectangular Grids):** For `(n_districts, n_weeks)`, summing across weeks per district collapses columns (`axis=1`), whereas averaging across districts per week collapses rows (`axis=0`). Testing on a non-square `(3, 4)` matrix catches swapped axes immediately.
- **Task 16.3 (Sigmoid Sign & Integer Cast):** Writing `np.exp(z)` instead of `np.exp(-z)` flips high-risk scores toward `0.0`, and forgetting `.astype(int)` returns boolean `[True, False]` instead of ML-ready `[1, 0]` labels.
