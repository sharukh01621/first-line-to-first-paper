## Exam Task 31.1 (30 Points): `temporal_district_window_split(epi_df, district, seq_len=8, cutoff_year=2025)`

Build a leakage-free chronological sliding-window split for a single district in `epi_surveillance` (without mutating `epi_df`):
1. Filter `epi_df[epi_df["district"] == district]` and sort chronologically by `["year", "epi_week"]`.
2. Extract numeric features in exact column order `["dengue_cases", "rainfall_mm", "mean_temp_c"]` (index `0` is the target `dengue_cases`).
3. Fit `StandardScaler()` **strictly on rows where `year < cutoff_year`**, then transform the full sorted feature matrix `scaled = scaler.transform(raw)`.
4. Build standardized sliding windows (both inputs and targets on the training-fitted `StandardScaler` scale):
   - **Train windows** strictly inside `year < cutoff_year`: for `t` in `range(seq_len, n_train)`, `X_train` window is `scaled[t - seq_len : t]` and `y_train` is `scaled[t, 0]`.
   - **Test windows** for `year >= cutoff_year`: for `t` in `range(n_train, len(sub))`, `X_test` window is `scaled[t - seq_len : t]` (strictly preceding history!) and `y_test` is `scaled[t, 0]`.
5. Return `{"X_train": X_train, "y_train": y_train, "X_test": X_test, "y_test": y_test, "scaler": scaler}`.

```python
def temporal_district_window_split(
    epi_df, district, seq_len=8, cutoff_year=2025
):
    """Leak-free temporal train/test sliding windows (standardized)."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    sub = (
        epi_df[epi_df["district"] == district]
        .sort_values(["year", "epi_week"])
        .reset_index(drop=True)
    )
    feat_cols = ["dengue_cases", "rainfall_mm", "mean_temp_c"]
    raw = sub[feat_cols].to_numpy(dtype=np.float32)
    n_train = int((sub["year"] < cutoff_year).sum())
    scaler = StandardScaler()
    scaler.fit(raw[:n_train])
    scaled = scaler.transform(raw).astype(np.float32)
    X_train = np.stack(
        [scaled[t - seq_len : t] for t in range(seq_len, n_train)]
    )
    y_train = scaled[seq_len:n_train, 0].astype(np.float32)
    X_test = np.stack(
        [scaled[t - seq_len : t] for t in range(n_train, len(sub))]
    )
    y_test = scaled[n_train:, 0].astype(np.float32)
    return {
        "X_train": X_train,
        "y_train": y_train,
        "X_test": X_test,
        "y_test": y_test,
        "scaler": scaler,
    }
    ### END SOLUTION
```

## Exam Task 31.2 (35 Points): `train_epi_gru_forecaster(X_train, y_train, X_test, y_test, hidden_size=16, epochs=40, lr=0.02, seed=42)`

Train a PyTorch GRU forecaster cleanly with gradient clipping:
1. Seed PyTorch via `torch.manual_seed(seed)` and construct a GRU model (`nn.GRU(input_size, hidden_size, batch_first=True)` + `nn.Linear(hidden_size, 1)` reading `out[:, -1, :]`).
2. Train for `epochs` epochs with `nn.MSELoss()`, `torch.optim.Adam(model.parameters(), lr=lr)`, and `nn.utils.clip_grad_norm_(model.parameters(), 1.0)` before each `optimizer.step()`.
3. Switch to `model.eval()` and `with torch.no_grad():` to compute 1D float NumPy predictions `preds` on `X_test`.
4. Return `{"test_rmse": round4, "test_mae": round4, "final_train_loss": round5, "preds": preds}`.

```python
class _ExamGRU(nn.Module):
    def __init__(self, input_size, hidden_size=16):
        super().__init__()
        self.gru = nn.GRU(input_size, hidden_size, batch_first=True)
        self.head = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        return self.head(out[:, -1, :])

def train_epi_gru_forecaster(
    X_train, y_train, X_test, y_test,
    hidden_size=16, epochs=40, lr=0.02, seed=42,
):
    """Train GRU forecaster with MSELoss, Adam & gradient clipping."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    torch.manual_seed(seed)
    x_tr = torch.tensor(np.asarray(X_train, dtype=np.float32))
    y_tr = torch.tensor(
        np.asarray(y_train, dtype=np.float32)
    ).unsqueeze(-1)
    x_te = torch.tensor(np.asarray(X_test, dtype=np.float32))
    y_te = np.asarray(y_test, dtype=float)
    model = _ExamGRU(x_tr.shape[-1], hidden_size=hidden_size)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.MSELoss()
    model.train()
    final_loss = 0.0
    for _ in range(epochs):
        opt.zero_grad()
        loss = loss_fn(model(x_tr), y_tr)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
        final_loss = float(loss.item())
    model.eval()
    with torch.no_grad():
        preds = model(x_te).squeeze(-1).numpy().astype(float)
    test_rmse = round(float(np.sqrt(np.mean((y_te - preds) ** 2))), 4)
    test_mae = round(float(np.mean(np.abs(y_te - preds))), 4)
    return {
        "test_rmse": test_rmse,
        "test_mae": test_mae,
        "final_train_loss": round(final_loss, 5),
        "preds": preds,
    }
    ### END SOLUTION
```

## Exam Task 31.3 (35 Points): `compare_gru_lstm_persistence_epi(epi_df, district="Dhaka", seq_len=8, seed=42)`

Run a complete leak-free epidemiological sequence benchmark (without mutating `epi_df`):
1. Call `temporal_district_window_split(epi_df, district=district, seq_len=seq_len, cutoff_year=2025)`.
2. Evaluate three models on the exact same held-out test set `(X_test, y_test)`:
   - `"Persistence"`: predicts `X_test[:, -1, 0]` (previous week's standardized dengue cases).
   - `"GRU"`: `train_epi_gru_forecaster(..., hidden_size=16, epochs=40, lr=0.02, seed=seed)`.
   - `"LSTM"`: same training setup (`hidden_size=16, epochs=40, lr=0.02, seed=seed`, `MSELoss`, `Adam`, `clip_grad_norm_(..., 1.0)`) using `nn.LSTM(..., batch_first=True)` + `nn.Linear(hidden_size, 1)`.
3. Build `leaderboard` as a `pd.DataFrame` with exact columns `["model", "test_rmse", "test_mae"]` (4 decimals), sorted ascending by `test_rmse` (`reset_index(drop=True)`).
4. Return `{"leaderboard": leaderboard, "best_model": str(leaderboard.loc[0, "model"])}`.

```python
class _ExamLSTM(nn.Module):
    def __init__(self, input_size, hidden_size=16):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.head = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.head(out[:, -1, :])

def compare_gru_lstm_persistence_epi(
    epi_df, district="Dhaka", seq_len=8, seed=42
):
    """Benchmark Persistence vs GRU vs LSTM on held-out test weeks."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    split = temporal_district_window_split(
        epi_df, district=district, seq_len=seq_len, cutoff_year=2025
    )
    X_tr, y_tr = split["X_train"], split["y_train"]
    X_te, y_te = split["X_test"], split["y_test"].astype(float)
    pers = X_te[:, -1, 0].astype(float)
    gru_res = train_epi_gru_forecaster(
        X_tr, y_tr, X_te, y_te,
        hidden_size=16, epochs=40, lr=0.02, seed=seed,
    )
    torch.manual_seed(seed)
    lstm_model = _ExamLSTM(X_tr.shape[-1], hidden_size=16)
    opt = torch.optim.Adam(lstm_model.parameters(), lr=0.02)
    loss_fn = nn.MSELoss()
    xt = torch.tensor(X_tr, dtype=torch.float32)
    yt = torch.tensor(y_tr, dtype=torch.float32).unsqueeze(-1)
    xe = torch.tensor(X_te, dtype=torch.float32)
    lstm_model.train()
    for _ in range(40):
        opt.zero_grad()
        loss = loss_fn(lstm_model(xt), yt)
        loss.backward()
        nn.utils.clip_grad_norm_(lstm_model.parameters(), 1.0)
        opt.step()
    lstm_model.eval()
    with torch.no_grad():
        lstm_preds = lstm_model(xe).squeeze(-1).numpy().astype(float)
    rows = [
        {
            "model": "Persistence",
            "test_rmse": round(
                float(np.sqrt(np.mean((y_te - pers) ** 2))), 4
            ),
            "test_mae": round(float(np.mean(np.abs(y_te - pers))), 4),
        },
        {
            "model": "GRU",
            "test_rmse": gru_res["test_rmse"],
            "test_mae": gru_res["test_mae"],
        },
        {
            "model": "LSTM",
            "test_rmse": round(
                float(np.sqrt(np.mean((y_te - lstm_preds) ** 2))), 4
            ),
            "test_mae": round(
                float(np.mean(np.abs(y_te - lstm_preds))), 4
            ),
        },
    ]
    lb = (
        pd.DataFrame(rows, columns=["model", "test_rmse", "test_mae"])
        .sort_values("test_rmse")
        .reset_index(drop=True)
    )
    return {"leaderboard": lb, "best_model": str(lb.loc[0, "model"])}
    ### END SOLUTION
```

## Chapter 31 Exam Debugging Post-Mortem (Solution Notebook Only)

- **Task 31.1 (Sorting & Training-Only Scaling):** Never trust input row order — sorting by `["year", "epi_week"]` protects against shuffled surveillance tables, and fitting `StandardScaler` exclusively on `year < cutoff_year` prevents 2025 monsoon peaks from leaking into 2024 training z-scores.
- **Task 31.2 (Gradient Clipping & Eval Hygiene):** Always place `nn.utils.clip_grad_norm_(model.parameters(), 1.0)` *after* `loss.backward()` and *before* `optimizer.step()`, and switch to `model.eval()` + `torch.no_grad()` when generating `preds`.
- **Task 31.3 (Apples-to-Apples Temporal Benchmarking):** Evaluating Persistence (`X_test[:, -1, 0]`), GRU, and LSTM on the exact same 52 held-out 2025 weeks proves that gated recurrence reduces dengue forecast RMSE by 11–18% on Dhaka and ~25% on Chattogram.
