## To-Do 27.1 (ReLU & Subgradient): `relu_and_grad(z)`

Implement `relu_and_grad(z)` without mutating `z`:
- Convert `z` to a float NumPy array (`np.asarray(z, dtype=float)`).
- Return `(np.maximum(0.0, z_arr), (z_arr > 0).astype(float))`.

```python
def relu_and_grad(z):
    """Return (ReLU(z), dReLU/dz) as float ndarrays without mutating z."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    z_arr = np.asarray(z, dtype=float)
    out = np.maximum(0.0, z_arr)
    grad = (z_arr > 0).astype(float)
    return (out, grad)
    ### END SOLUTION
```

## To-Do 27.2 (Numerically Stable Softmax): `stable_softmax(logits)`

Implement overflow-safe `stable_softmax(logits)` for a 2D `(n, k)` array without mutating `logits`:
- Subtract the row-wise maximum `np.max(z, axis=1, keepdims=True)` before exponentiating so extreme logits like `[1000.0, 1001.0, 999.0]` never overflow.
- Divide by row sums (`axis=1, keepdims=True`) so every row sums to `1.0`.

```python
def stable_softmax(logits):
    """Compute overflow-safe row-wise softmax on a 2D (n, k) array."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    z = np.asarray(logits, dtype=float)
    shifted = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(shifted)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
    ### END SOLUTION
```

## To-Do 27.3 (Binary Cross-Entropy & Logit Gradient): `binary_bce_loss_and_grad(y_true, logits)`

Implement `binary_bce_loss_and_grad(y_true, logits)` without mutating inputs:
- Compute clipped sigmoid probabilities `p = np.clip(1 / (1 + np.exp(-np.clip(z, -500, 500))), 1e-12, 1 - 1e-12)`.
- Compute mean BCE loss rounded to **5 decimal places** (`float`) and analytical gradient `dlogits = (p - y_true) / len(y_true)`.
- Return `(loss_round5, dlogits_array)`.

```python
def binary_bce_loss_and_grad(y_true, logits):
    """Return (mean_bce_round5, dlogits) from binary labels & logits."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    y = np.asarray(y_true, dtype=float)
    z = np.asarray(logits, dtype=float)
    p_raw = 1.0 / (1.0 + np.exp(-np.clip(z, -500.0, 500.0)))
    p = np.clip(p_raw, 1e-12, 1.0 - 1e-12)
    loss = -float(np.mean(
        y * np.log(p) + (1.0 - y) * np.log(1.0 - p)
    ))
    dlogits = (p - y) / len(y)
    return (round(loss, 5), dlogits)
    ### END SOLUTION
```

## To-Do 27.4 (Two-Layer MLP Forward Pass): `mlp_forward_two_layer(X, W1, b1, W2, b2)`

Implement the forward pass of a 2-layer ReLU-to-Sigmoid MLP without mutating inputs:
- Compute `Z1 = X @ W1 + b1`, `A1 = np.maximum(0.0, Z1)`, `Z2 = A1 @ W2 + b2`, and `probs = 1 / (1 + np.exp(-np.clip(Z2, -500, 500)))`.
- Return `(probs, cache)` where `cache` has exact keys `{"X", "Z1", "A1", "Z2", "probs"}`.

```python
def mlp_forward_two_layer(X, W1, b1, W2, b2):
    """Run 2-layer ReLU->Sigmoid forward pass; return (probs, cache)."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    X_a = np.asarray(X, dtype=float)
    Z1 = X_a @ W1 + b1
    A1 = np.maximum(0.0, Z1)
    Z2 = A1 @ W2 + b2
    probs = 1.0 / (1.0 + np.exp(-np.clip(Z2, -500.0, 500.0)))
    cache = {
        "X": X_a, "Z1": Z1, "A1": A1,
        "Z2": Z2, "probs": probs,
    }
    return (probs, cache)
    ### END SOLUTION
```

## To-Do 27.5 (Analytical Two-Layer Backprop): `mlp_backward_two_layer(y_true, cache, W2)`

Implement analytical reverse-mode backpropagation from mean BCE loss:
- Reshape `y_col = np.asarray(y_true, dtype=float).reshape(-1, 1)` to prevent `(n, 1) - (n,)` broadcasting bugs.
- Compute `dZ2 = (probs - y_col) / n`, `dW2`, `db2` (`keepdims=True`), `dZ1 = (dZ2 @ W2.T) * (Z1 > 0)`, `dW1`, `db1` (`keepdims=True`).
- Return `{"dW1": dW1, "db1": db1, "dW2": dW2, "db2": db2}` matching finite-difference gradients within `1e-5`.

```python
def mlp_backward_two_layer(y_true, cache, W2):
    """Compute analytical BCE backprop gradients for a 2-layer MLP."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    y_col = np.asarray(y_true, dtype=float).reshape(-1, 1)
    n = len(y_col)
    dZ2 = (cache["probs"] - y_col) / n
    dW2 = cache["A1"].T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)
    dA1 = dZ2 @ np.asarray(W2, dtype=float).T
    dZ1 = dA1 * (cache["Z1"] > 0).astype(float)
    dW1 = cache["X"].T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)
    return {"dW1": dW1, "db1": db1, "dW2": dW2, "db2": db2}
    ### END SOLUTION
```

## To-Do 27.6 (Centered Finite-Difference Check): `numerical_gradient_check(loss_fn, W, eps=1e-5)`

Implement centered finite differences on a `.copy()` of `W` without mutating `W`:
- For each multi-index `idx` in `np.ndindex(W_work.shape)`, evaluate `(loss_fn(W_plus) - loss_fn(W_minus)) / (2 * eps)` and restore `W_work[idx]`.
- Return the numerical gradient ndarray with the exact same shape as `W`.

```python
def numerical_gradient_check(loss_fn, W, eps=1e-5):
    """Compute centered finite-difference gradient without mutating W."""
    # STUB: raise NotImplementedError("Replace with your solution")
    ### BEGIN SOLUTION
    W_work = np.asarray(W, dtype=float).copy()
    grad = np.zeros_like(W_work)
    for idx in np.ndindex(W_work.shape):
        orig = W_work[idx]
        W_work[idx] = orig + eps
        lp = float(loss_fn(W_work))
        W_work[idx] = orig - eps
        lm = float(loss_fn(W_work))
        W_work[idx] = orig
        grad[idx] = (lp - lm) / (2.0 * eps)
    return grad
    ### END SOLUTION
```

## Annotated Debugging Post-Mortem (Solution Notebook Only)

- **To-Do 27.2 (Forgetting `keepdims=True` in Softmax):** Calling `z - np.max(z, axis=1)` subtracts a 1D `(n,)` vector across the last axis, causing either a `ValueError` or silent row/column corruption when `n == k`. Always pass `axis=1, keepdims=True`.
- **To-Do 27.3 (Forgetting `/ len(y)` or Probability Clipping):** Mean BCE divides the sum by `n`, so its gradient is `(p - y) / n` (not `p - y`). Clipping probabilities to `[1e-12, 1 - 1e-12]` guarantees `np.log(0.0)` never produces `-inf` or `nan`.
- **To-Do 27.5 (`(n, 1) - (n,)` Broadcasting Explosion):** Subtracting a 1D target `y_true` of shape `(n,)` from `probs` of shape `(n, 1)` silently creates an `(n, n)` outer-difference matrix! Always reshape `y_col = np.asarray(y_true, dtype=float).reshape(-1, 1)` on line 1 of backprop.
